# **Denoising microscope**

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def gaussion_kernal(size,size_y = None):
    size = int(size)
    if not size_y:
        size_y = size
    else:
        size_y = int(size)
        x,y = np.mgrid[-size:size+1,-size_y:size_y+1]
        g = np.exp(-(x**2/float(size)+y**2/float(size_y)))
        return g / g.sum()
```

```python
gausian_kerna_Array=gaussion_kernal(3)
print(gausian_kerna_Array)
plt.imshow(gausian_kerna_Array,cmap = plt.get_cmap('jet'),interpolation = 'nearst')
plt.colorbar()
plt.show()
```

```python
from skimage import io,img_as_float
from scipy import ndimage as nd
```

```python
img = img_as_float(io.imread(""))
gaussin_kernal = nd.gaussian_filter(img,sigma =3)
plt.imshow(gaussin_kernal)
```

```python
median_img = nd.median_filter(img,size=3)
plt.imshow(median_img)
```

```python
from skimage.restoration import denoise_nl_means,estimate_sigma
sigma_est = np.mean(estimate_sigma(img,multichannrl = True))
denoise = denoise_nl_means(img,h=1.5 * sigma_est,fast_mode = False,patch_size = 5,patch_distance = 3,multichannel = True)
plt.imshow()
```
