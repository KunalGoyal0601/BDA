```python
img = img_as_ubyte(io.imread(r"E:\BDA\digital image processing\Practice_images\lena_image.png",as_gray=True))
print(img.dtype,img.shape)
```

```python
print(img[0,0])# a pixal value
print(img[10:15,20:25])# a range of pixal value
```

```python
min_value = img.min()
max_value = img.max()
mean_gray = img.mean()
print("Min,Max and Mean are:",min_value,max_value,mean_gray)
```

# flipping image

```python
filppedLR = np.fliplr(img)
flippedUD = np.flipud(img)
plt.subplot(2,1,1)
plt.imshow(img,cmap="Greys")
plt.subplot(2,2,3)
plt.imshow(filppedLR)
plt.subplot(2,2,4)
plt.imshow(flippedUD)
```

# rotation

```python
rotation = ndimage.rotate(img,45,reshape=False)
plt.imshow(rotation)
```

```python
rotation = ndimage.rotate(img,45)
plt.imshow(rotation)
```

# filteration

```python
uniform_filterd = ndimage.uniform_filter(img,size = 9)#bluring filter
plt.imshow(uniform_filterd)
```

```python
gausiian_filter = ndimage.gaussian_filter(img,sigma=7)#noice smoothing sigma high bluer high
plt.imshow(gausiian_filter)
```

```python
median_filter = ndimage.median_filter(img,7)
plt.imshow(median_filter)
```

```python
sobel_img = ndimage.sobel(img)
plt.imshow(sobel_img)
```
