# **Histogram based segmrntation**

```python
from skimage import io,img_as_float,img_as_ubyte
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = img_as_float(io.imread(" "))
```

```python
from skimage.restoration import denoise_nl_means,estimate_sigma
sigma_est = np.mean(estimate_sigma(img,multichannrl = True))
denoise = denoise_nl_means(img,h=1.5 * sigma_est,fast_mode = True,patch_size = 5,patch_distance = 3,multichannel = True)
denoice_ubyte = img_as_ubyte(denoise)
plt.imshow(denoice_ubyte,cmap = "gray")
```

```python
plt.hist(denoice_ubyte.flat,bins = 100 , range = (0,255))
```

```python
segm1 = (denoice_ubyte <= 55)
segm2 = (denoice_ubyte > 55) & (denoice_ubyte <= 100)
segm3 = (denoice_ubyte > 110) & (denoice_ubyte <= 120)
segm4 = (denoice_ubyte > 210)

all_segments = np.zeros((denoice_ubyte.shape[0],denoice_ubyte.shape[1],3))
all_segments[segm1] = (1,0,0)
all_segments[segm1] = (0,1,0)
all_segments[segm1] = (0,0,1)
all_segments[segm1] = (1,1,0)
plt.imshow(all_segments)
```

```python
from scipy import ndimage as nd
segm1_opened = nd.binary_opening(segm1,np.ones((3,3)))
segm1_closed = nd.binary_closing(segm1_opened,np.ones((3,3)))

segm2_opened = nd.binary_opening(segm2,np.ones((3,3)))
segm2_closed = nd.binary_closing(segm2_opened,np.ones((3,3)))

segm3_opened = nd.binary_opening(segm3,np.ones((3,3)))
segm3_closed = nd.binary_closing(segm3_opened,np.ones((3,3)))

segm4_opened = nd.binary_opening(segm4,np.ones((3,3)))
segm4_closed = nd.binary_closing(segm4_opened,np.ones((3,3)))

all_segments_cleaned = np.zeros((denoice_ubyte.shape[0],denoice_ubyte.shape[1],3))
all_segments_cleaned[segm1_closed] = (1,0,0)
all_segments_cleaned[segm2_closed] = (0,1,0)
all_segments_cleaned[segm3_closed] = (0,0,1)
all_segments_cleaned[segm4_closed] = (1,1,0)

plt.imshow(all_segments_cleaned)
```
