# **Read multiple files Glob**

```python
import cv2
import glob
```

```python
path = r"E:\BDA\digital image processing\Practice_images\*"#we put \* to see every file in the folder
for file in glob.glob(path):
    print(file)
    a = cv2.imread(file)
    cv2.imshow("original_img",a)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
    
    c = cv2.cvtColor(a,cv2.COLOR_BGR2RGB)
    cv2.imshow("color_img",c)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
