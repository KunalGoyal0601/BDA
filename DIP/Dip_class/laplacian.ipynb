# **Edge detection through Laplacian Operator**

image

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Read image in grayscale
img = cv2.imread(r"C:\BDA\digital image processing\Dip_class\dip_img\lena_image.png", cv2.IMREAD_GRAYSCALE)

# Convert to int32 to avoid overflow
img = img.astype(np.int32)

rows, cols = img.shape

# Initialize Laplacian output
laplacian = np.zeros_like(img)

# Apply Laplacian filter manually
for i in range(1, rows - 1):
    for j in range(1, cols - 1):
        laplacian[i, j] = (
            img[i + 1, j] - 2 * img[i, j] + img[i - 1, j] +
            img[i, j + 1] - 2 * img[i, j] + img[i, j - 1]
        )

# Take absolute and clip values
laplacian = np.abs(laplacian)
laplacian = np.uint8(np.clip(laplacian, 0, 255))

# Display images
plt.figure(figsize=(20, 10))

plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(img, cmap='gray')

plt.subplot(1, 2, 2)
plt.title('Laplacian Edge')
plt.imshow(laplacian, cmap='gray')

plt.show()
```

matrix

```python
import numpy as np

# Given matrix
img = np.array([
    [10, 10, 10, 10, 10],
    [10, 10, 10, 10, 10],
    [10, 10, 10, 10, 10],
    [200, 200, 200, 200, 200],
    [200, 200, 200, 200, 200]
])

print("given matrix:\n", img)

r, c = img.shape

# Method 1: using zeros_like
L = np.zeros_like(img)

for i in range(1, r - 1):
    for j in range(1, c - 1):
        L[i, j] = (
            img[i + 1, j] - 2 * img[i, j] + img[i - 1, j] +
            img[i, j + 1] - 2 * img[i, j] + img[i, j - 1]
        )

print("\nLaplacian_out_matrix:\n", L)
```

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Original Matrix')
plt.imshow(img, cmap='gray')
plt.subplot(1, 2, 2)
plt.title('Laplacian Output Matrix')
plt.imshow(L, cmap='gray')
plt.show()
```
