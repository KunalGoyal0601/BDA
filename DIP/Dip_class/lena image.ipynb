```python
import cv2
import matplotlib.pyplot as plt

```

```python
# Reade the image
image = cv2.imread(r"C:\BDA\digital image processing\Dip_class\dip_img\lena_image.png")

#display the image 
plt.imshow(image);
```

# **Display the image features**

```python
#height, width, number of channel
height, width, channel = image.shape
print("Height:", height)
print("Width:", width)
print("Number of channels:", channel)
```

```python
#chack if it is a grayscale image
if channel == 1:
    print("The image is grayscale")
    
else:
    print("The image is colored")    
```

```python
#convert the colored image into grayscale image
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
plt.imshow(gray_image, cmap='gray');
```

```python
#display histogram of the gray scale image
hist = cv2.calcHist([gray_image], [0], None, [256], [0, 256])
plt.plot(hist);
```

```python
# disply histogram of colored image
colors = ('b', 'g', 'r')
for i, color in enumerate(colors):
    hist = cv2.calcHist([image], [i], None, [256], [0, 256])
    plt.plot(hist, color=color)
```
