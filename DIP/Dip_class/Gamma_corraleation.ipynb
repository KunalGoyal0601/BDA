# **GAMMA CORRECTION TECHNIQUE**

matrix

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2

I = np.array([[50, 100],
              [150, 200]], dtype=np.float32)
I_norm = I / 255.0
gamma = 1
I_gamma = np.power(I_norm, gamma)
I_final = np.uint8(I_gamma * 255)
print("Original Image:\n", I)
print("After Gamma Correction:\n", I_final)
```

```python
plt.subplot(1,2,1)
plt.imshow(I, cmap='gray', vmin=0, vmax=255)
plt.title("Original Image")
plt.colorbar()
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(I_final, cmap='gray', vmin=0, vmax=255)
plt.title("Gamma Corrected Image")
plt.colorbar()
plt.axis('off')

plt.show()
```

```python
img = cv2.imread(r"C:\BDA\digital image processing\Dip_class\dip_img\lena_image.png")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
def gamma_correction(channel, gamma):
    rows, cols = channel.shape
    corrected = np.zeros((rows, cols), dtype=np.uint8)

    for i in range(rows):
        for j in range(cols):
            r = channel[i, j] / 255.0
            s = (r ** gamma) * 255
            corrected[i, j] = s

    return corrected
```

```python
R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]
```

```python
gamma = 4

R_g = gamma_correction(R, gamma)
G_g = gamma_correction(G, gamma)
B_g = gamma_correction(B, gamma)
```

```python
gamma_img = np.stack((R_g, G_g, B_g), axis=2)
```

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.title("Original Color Image")
plt.imshow(img)
plt.axis("off")

plt.subplot(1,2,2)
plt.title(f"Gamma Corrected Image (γ = {gamma})")
plt.imshow(gamma_img)
plt.axis("off")

plt.show()
```
