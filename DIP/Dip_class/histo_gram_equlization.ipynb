```python
import numpy as np
import matplotlib.pyplot as plt
import cv2 
```

```python
#reading image
img = cv2.imread("E:\digital image processing\Sun_image_nasa.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img)
```

```python
img_1 = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
plt.imshow(img_1, cmap="gray")
```

```python
# setting grid_size
plt.figure(figsize=(10,12))
#plotting original image
plt.subplot(2,2,1)
plt.title("Original Image")
plt.imshow(img)

#plotting histogram of original image
img_hist = cv2.calcHist([img_1], [0], None, [256], [0,256])
plt.subplot(2,2,2)
plt.title("Histogram of Original Image")
plt.plot(img_hist)

#plotting histogram using ravel function
plt.subplot(2,2,3)
plt.hist(img_1.ravel(),256,[0,256])
plt.title("Histogram using ravel function")

# plotting histogram for colors in the image
for i, col in enumerate(["r","g","b"]):
    hist = cv2.calcHist([img], [i], None, [256], [0,256])
    plt.subplot(2,2,4)
    plt.plot(hist, color=col)
    plt.title("Histogram for colors in the image")
```

```python
#plotting equalized image
img_equ = cv2.equalizeHist(img_1)
plt.subplot(2,2,3)
plt.title("Equalized Image")
plt.imshow(img_equ, cmap="gray")
#plotting histogram of equalized image
img_equ_hist = cv2.calcHist([img_equ], [0], None, [256], [0,256])
plt.subplot(2,2,4)
plt.title("Histogram of Equalized Image")
plt.plot(img_equ_hist)
```
