# **Contrast Stretching Technique**
---

**I(enhenced(x,y)) = [I(x,y)] - min(I)/max(I) - min(I)] * 255**

```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

**Identify min(I) and max(I)**
1. method 1

```python
# take image
I = np.array([[50,100,150],
              [200,75,125],
              [175,25,225]])
min_ = I.min()
max_ = I.max()
print(min_,max_)
```

**Apply constrast streaching**

```python
Ienhenced = [np.round((i - min_) / (max_ - min_) * 255).astype(np.uint8) for i in I]
print(Ienhenced)
```

```python
Ienhenced = [((i - min_)/(max_ - min_)*255) for i in I]
Ienhenced = np.round(Ienhenced).astype(np.uint8)
print(Ienhenced)
```

```python
plt.imshow(Ienhenced)
plt.show()
```

# **With Image**

```python
import cv2
from matplotlib import pyplot as plt
import numpy as np

# Read grayscale image
img = cv2.imread(r"C:\BDA\digital image processing\Dip_class\dip_img\stretch_original.webp", cv2.IMREAD_GRAYSCALE)

# Convert to float for processing
img_float = img.astype(np.float32)

# Find min and max pixel values
I_min = np.min(img_float)
I_max = np.max(img_float)

# Apply contrast stretching
stretched = (img_float - I_min) / (I_max - I_min) * 255

# Convert back to uint8
stretched = np.uint8(stretched)

# Display
print("Original Image:")
plt.imshow(img, cmap='gray')
plt.show()
print("Contrast Stretched Image:")
plt.imshow(stretched, cmap='gray')
plt.show()
```
