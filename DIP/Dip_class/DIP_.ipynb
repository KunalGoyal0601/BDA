## Contrast Stretching Techniques.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

I=[[50,100,150],
    [200,75,125],
    [175,25,225]]
img=np.array(I)

# Find min and max pixel values
r_min = np.min(img)
r_max = np.max(img)

# Create output image
stretched = np.zeros(img.shape, dtype=np.uint8)

# Apply contrast stretching
for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        stretched[i, j] = ((img[i, j] - r_min) / (r_max - r_min)) * 255

# Display
plt.figure(figsize=(10,8))
plt.subplot(1,2,1), plt.imshow(img, cmap='gray'), plt.title('Original')
plt.subplot(1,2,2), plt.imshow(stretched, cmap='gray'), plt.title('Contrast Stretched')
plt.show()

print("Orignal image array: ",img)
print("\nContrast stretched image array: ",stretched)
```

## Gamma Correction Techniques

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

I=[[50,100],
    [150,200]]
img=np.array(I)

gamma = 2.0  # change value (0.5, 1.5, 2.0)

# Normalize image to 0-1
img_norm = img / 255.0

# Apply gamma correction
gamma_corrected = np.power(img_norm, gamma)

# Scale back to 0-255
gamma_corrected = np.uint8(gamma_corrected * 255)

# Display
plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title('Original')

plt.subplot(1,2,2)
plt.imshow(gamma_corrected, cmap='gray')
plt.title(f'Gamma = {gamma}')
plt.show()

print(img)
print(gamma_corrected)
```

```python
# Contrast Stretching, Gamma Correction and Histogram Equalization
# Digital Image Processing Experiment

import cv2
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Read Grayscale Image
# -----------------------------
img = cv2.imread('lena.jpeg', 0)   # Replace with your image path

# -----------------------------
# 1. Contrast Stretching
# -----------------------------
min_val = np.min(img)
max_val = np.max(img)

contrast = ((img - min_val) / (max_val - min_val)) * 255
contrast = contrast.astype(np.uint8)

# -----------------------------
# 2. Gamma Correction
# -----------------------------
gamma = 0.5   # change value (>1 darker, <1 brighter)

gamma_img = np.array(255 * (img / 255) ** gamma, dtype='uint8')

# -----------------------------
# 3. Histogram Equalization
# -----------------------------
hist_eq = cv2.equalizeHist(img)

# =====================================================
# Show All Images in ONE Figure (Subplot)
# =====================================================
plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')



plt.subplot(2,2,2)
plt.imshow(contrast, cmap='gray')
plt.title("Contrast Stretching")
plt.axis('off')

plt.subplot(2,2,3)
plt.imshow(gamma_img, cmap='gray')
plt.title("Gamma Correction")
plt.axis('off')

plt.subplot(2,2,4)
plt.imshow(hist_eq, cmap='gray')
plt.title("Histogram Equalization")
plt.axis('off')

plt.tight_layout()
plt.show()

# =====================================================
# Histograms (Separate Figures)
# =====================================================

plt.figure()
plt.title("Histogram - Original")
plt.hist(img.ravel(), bins=256)
plt.show()

plt.figure()
plt.title("Histogram - Contrast Stretching")
plt.hist(contrast.ravel(), bins=256)
plt.show()

plt.figure()
plt.title("Histogram - Gamma Correction")
plt.hist(gamma_img.ravel(), bins=256)
plt.show()

plt.figure()
plt.title("Histogram - Histogram Equalization")
plt.hist(hist_eq.ravel(), bins=256)
plt.show()
```

```python
# W.A.P to count Sun Spots using Digital Image Processing

import cv2
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Step 1: Read Image
# -----------------------------
img = cv2.imread('sun_big_spot.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# -----------------------------
# Step 2: Blur (Remove Noise)
# -----------------------------
blur = cv2.GaussianBlur(gray, (5,5), 0)

# -----------------------------
# Step 3: Thresholding
# (Sun spots are dark areas)
# -----------------------------
_, thresh = cv2.threshold(blur, 60, 255, cv2.THRESH_BINARY_INV)

# -----------------------------
# Step 4: Morphological Operation
# -----------------------------
kernel = np.ones((3,3), np.uint8)
opening = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)

# -----------------------------
# Step 5: Find Contours
# -----------------------------
contours, _ = cv2.findContours(opening, 
                               cv2.RETR_EXTERNAL, 
                               cv2.CHAIN_APPROX_SIMPLE)

# Filter small areas (remove noise)
spot_count = 0
for cnt in contours:
    area = cv2.contourArea(cnt)
    if area > 50:   # adjust threshold if needed
        spot_count += 1
        cv2.drawContours(img, [cnt], -1, (0,255,0), 2)

print("Number of Sun Spots Detected:", spot_count)

# -----------------------------
# Show Results
# -----------------------------
plt.figure()
plt.title("Detected Sun Spots")
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
# W.A.P to detect plant disease using Digital Image Processing

import cv2
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Step 1: Read Image
# -----------------------------
img = cv2.imread('plant_diseas.jpg') 
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# -----------------------------
# Step 2: Convert to HSV
# -----------------------------
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# -----------------------------
# Step 3: Define Disease Color Range
# (Brown/Yellow infected areas)
# -----------------------------
lower = np.array([10, 50, 50])
upper = np.array([35, 255, 255])

mask = cv2.inRange(hsv, lower, upper)

# -----------------------------
# Step 4: Remove Noise
# -----------------------------
kernel = np.ones((5,5), np.uint8)
mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)

# -----------------------------
# Step 5: Highlight Diseased Area
# -----------------------------
result = img_rgb.copy()
result[mask != 0] = [255, 0, 0]   # Mark infected area in red

# -----------------------------
# Step 6: Calculate Disease %
# -----------------------------
infected_pixels = np.sum(mask > 0)
total_pixels = mask.size
percentage = (infected_pixels / total_pixels) * 100

print("Disease Affected Area: {:.2f}%".format(percentage))

# -----------------------------
# Display Results
# -----------------------------
plt.figure(figsize=(10,5))

plt.subplot(1,3,1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(mask, cmap='gray')
plt.title("Disease Mask")
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(result)
plt.title("Detected Disease")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# W.A.P to Detect Diabetic Retinopathy using DIP

import cv2
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Step 1: Read Retina Image
# -----------------------------
img = cv2.imread('Dibatic_retinopathy.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# -----------------------------
# Step 2: Extract Green Channel
# (Best for blood vessel detection)
# -----------------------------
green = img[:,:,1]

# -----------------------------
# Step 3: Apply CLAHE
# (Contrast Enhancement)
# -----------------------------
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
enhanced = clahe.apply(green)

# -----------------------------
# Step 4: Gaussian Blur
# -----------------------------
blur = cv2.GaussianBlur(enhanced, (5,5), 0)

# -----------------------------
# Step 5: Thresholding
# (Detect bright lesions / exudates)
# -----------------------------
_, thresh = cv2.threshold(blur, 180, 255, cv2.THRESH_BINARY)

# -----------------------------
# Step 6: Morphological Operation
# -----------------------------
kernel = np.ones((3,3), np.uint8)
opening = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel)

# -----------------------------
# Step 7: Find Abnormal Regions
# -----------------------------
contours, _ = cv2.findContours(opening, 
                               cv2.RETR_EXTERNAL, 
                               cv2.CHAIN_APPROX_SIMPLE)

lesion_count = 0
for cnt in contours:
    area = cv2.contourArea(cnt)
    if area > 30:
        lesion_count += 1
        cv2.drawContours(img_rgb, [cnt], -1, (255,0,0), 2)

# -----------------------------
# Result
# -----------------------------
print("Possible Lesions Detected:", lesion_count)

if lesion_count > 5:
    print("Diabetic Retinopathy Detected (Possible Case)")
else:
    print("No Significant DR Signs Detected")

# -----------------------------
# Display Result
# -----------------------------
plt.figure(figsize=(8,6))
plt.imshow(img_rgb)
plt.title("Diabetic Retinopathy Detection Result")
plt.axis('off')
plt.show()
```

```python
# W.A.P to Enhance Thumb Image

import cv2
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------
# Step 1: Read Thumb Image
# -----------------------------
img = cv2.imread('thumb_image.jpg')   # Put your thumb image name here
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# -----------------------------
# Step 2: Convert to Grayscale
# -----------------------------
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# -----------------------------
# Step 3: Apply CLAHE (Contrast Enhancement)
# -----------------------------
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
enhanced = clahe.apply(gray)

# -----------------------------
# Step 4: Noise Removal
# -----------------------------
denoise = cv2.GaussianBlur(enhanced, (5,5), 0)

# -----------------------------
# Step 5: Sharpening
# -----------------------------
kernel = np.array([[0, -1, 0],
                   [-1, 5,-1],
                   [0, -1, 0]])

sharpen = cv2.filter2D(denoise, -1, kernel)

# -----------------------------
# Display Results
# -----------------------------
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(gray, cmap='gray')
plt.title("Original Thumb Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(sharpen, cmap='gray')
plt.title("Enhanced Thumb Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```

## Image filtering(Convolution process):

```python
import numpy as np
import matplotlib.pyplot as plt

# Input image (3x3 matrix)
image = np.array([[10, 20, 30],
                  [40, 50, 60],
                  [70, 80, 90]])

# 3x3 Average kernel
kernel = np.ones((3,3)) / 9

# Zero padding
padded = np.pad(image, 1, mode='constant', constant_values=0)

# Output image
output = np.zeros_like(image, dtype=float)

# Convolution process
for i in range(image.shape[0]):
    for j in range(image.shape[1]):

        # Extract 3x3 region
        region = padded[i:i+3, j:j+3]

        # Multiply and sum
        output[i,j] = np.sum(region * kernel)

# Display results
print("Original Image:\n", image)
print("\nKernel Image:\n", kernel)
print("\nPaddaed Image:\n", padded)
print("\nFiltered Image:\n", output.astype(int))

# Step 4: Display Images
# -----------------------------
plt.figure(figsize=(12,6))

# Original Image
plt.subplot(1,4,1)
plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.axis()

# Kernel Image
plt.subplot(1,4,2)
plt.imshow(kernel, cmap='gray')
plt.title("Kernel Image")
plt.axis()

# Padding Image
plt.subplot(1,4,3)
plt.imshow(padded, cmap='gray')
plt.title("padding Image")
plt.axis()

# Filtered Image
plt.subplot(1,4,4)
plt.imshow(output, cmap='gray')
plt.title("Filtered Image")
plt.axis()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ======================
# STEP 1: Load Image
# ======================
image = plt.imread("lena.jpeg")   # put your image name

# Convert RGB → Grayscale
if len(image.shape) == 3:
    image = np.mean(image, axis=2)

# Convert to float (important for calculations)
image = image.astype(float)

# ======================
# STEP 2: Average Kernel
# ======================
kernel = np.ones((3,3)) / 9

# ======================
# STEP 3: Zero Padding
# ======================
padded = np.pad(image, 1, mode='constant', constant_values=0)

# ======================
# STEP 4: Filtering (Convolution)
# ======================
output = np.zeros_like(image)

for i in range(image.shape[0]):
    for j in range(image.shape[1]):

        region = padded[i:i+3, j:j+3]
        output[i,j] = np.sum(region * kernel)

# ======================
# STEP 5: MSE Calculation
# ======================
mse = np.mean((image - output) ** 2)

# ======================
# STEP 6: PSNR Calculation
# ======================
MAX_PIXEL = 255.0

if mse == 0:
    psnr = float('inf')
else:
    psnr = 10 * np.log10((MAX_PIXEL**2) / mse)

# ======================
# STEP 7: Print Statistics
# ======================
print("MSE :", mse)
print("PSNR :", psnr, "dB")

# ======================
# STEP 8: Display Images
# ======================
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(output, cmap='gray')
plt.title("Filtered Image")
plt.axis('off')

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ======================
# STEP 1: Load Image
# ======================
image = plt.imread("lena.jpeg")

# Convert RGB → Grayscale
if len(image.shape) == 3:
    image = np.mean(image, axis=2)

image = image.astype(float)

# ======================
# STEP 2: SHARPEN KERNEL
# ======================
kernel = np.array([[0, -1, 0],
                   [-1, 5, -1],
                   [0, -1, 0]])

# ======================
# STEP 3: Zero Padding
# ======================
padded = np.pad(image, 1, mode='constant', constant_values=0)

# ======================
# STEP 4: Convolution
# ======================
output = np.zeros_like(image)

for i in range(image.shape[0]):
    for j in range(image.shape[1]):

        region = padded[i:i+3, j:j+3]
        output[i,j] = np.sum(region * kernel)

# Clip values (important)
output = np.clip(output, 0, 255)

# ======================
# STEP 5: Display
# ======================
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(image, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1,2,2)
plt.imshow(output, cmap='gray')
plt.title("Sharpened Image")
plt.axis('off')

plt.show()
```
