# **Image Display**
---

```python
# Import Necessary Libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
#step 1 define a simple 2*2 image
R = np.array([[255, 0],
         [128,64]]) 

G = np.array([[0,255],
     [64,128]])

B = np.array([[125,64],
     [255,0]]
)
print("Red Channel:",R)
print("Green Channel:",G)
print("Blue Channel:",B)
```

```python
#step 2 stack the 3 channels to create a color image display
import matplotlib.pyplot as plt
```

```python
plt.imshow(R, cmap='Reds')
plt.colorbar()
plt.title('Red Channel')
plt.show()
```

```python
#display the green channel
plt.imshow(G, cmap='Greens')
plt.colorbar()
plt.title('Green Channel')
plt.show()
```

```python
#display the blue channel
plt.imshow(B,cmap = "Blues")
plt.colorbar()
plt.title('Blue Channel')
plt.show()
```

```python
#1.different ways to stack the channels
img = np.dstack((R,G,B))
plt.imshow(img)
```

```python
#2 stack the 3 channels to create a color image
pixel_matrix = [[255, 0, 125],
                [0, 255, 64],
                [128, 64, 255],
                [64, 128, 0]]
color_image = np.array(pixel_matrix, dtype=np.uint8).reshape((2, 2, 3))
plt.imshow(color_image)
```

```python
#3.
p = np.zeros((3,3,3), dtype=np.uint8)
p[1,1] = [255,0,125]
p[1,2] = [0,255,64]
p[2,1] = [128,64,255]
p[2,2] = [64,128,0]
plt.imshow(p)
```

```python
#4.
a= [255,0,125]
b= [0,255,64]
c= [128,64,255]
d= [64,128,0]
p = np.array ([[a,b],
     [c,d]])
plt.imshow(p)
```

# **GREYSCALE IMAGE**
---

```python
#
BRAIN_TUMER = np.array([[255,252],
                       [251,250]])
plt.imshow(BRAIN_TUMER, cmap='gray')
```

```python
#convert this color image to gray scale image without using any library function
#RED CHANNEL MATRIX IN GREY_SCALE IMAGE
R = np.array([[255, 0],
            [128, 64]])

plt.imshow(R, cmap='gray')
plt.colorbar()
plt.title('RED CHANNEL')
plt.show()
```

```python
#GREEN CHANNEL MATRIX IN GREY_SCALE IMAGE
G = np.array([[0, 255],
            [64, 128]])

plt.imshow(G, cmap='gray')
plt.colorbar()
plt.title('GREEN CHANNEL')
plt.show()
```

```python
#BLUE CHANNEL MATRIX IN GREY_SCALE IMAGE
B = np.array([[128, 64],
            [255, 0]])

plt.imshow(B, cmap='gray')
plt.colorbar()
plt.title('BLUE CHANNEL')
plt.show()
```

```python
#COLOR_IMAGE
rgb_image = np.stack([R, G, B], axis=2)
```

```python
plt.imshow(rgb_image)
plt.axis('off')
plt.show()
```
