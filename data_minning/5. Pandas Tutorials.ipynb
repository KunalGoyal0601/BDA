### Creating a Data Frame

```python
# Load library
import pandas as pd
# Create DataFrame
dataframe = pd.DataFrame()
# Add columns
dataframe['Name'] = ['Jack', 'Steven ']
dataframe['Age'] = [38, 25]
dataframe['Driver'] = [True, False]
# Show DataFrame
dataframe
```

Alternatively, once we have created a DataFrame object, we can append new rows to the bottom:

```python
# Create row
new_person = pd.Series(['Bob', 40, True], index=['Name','Age','Driver'])
# Append row
dataframe.append(new_person, ignore_index=True)
```

### Navigating DataFrames

```python
# importing pandas as pd
import pandas as pd

# importing numpy as np
import numpy as np

# dictionary of lists
dict = {'Name':['A','B','C','D'],
        'Age':[25,22,28,22],
        'Sex':['Female','Male','Female','Male'],
        'First Score':[100, 90, np.nan, 95],
        'Second Score': [30, 45, 56, np.nan],
        'Third Score':[np.nan, 40, 80, 98]}

# creating a dataframe from dictionary
dataframe = pd.DataFrame(dict)
dataframe
```

```python
# Select first row
dataframe.iloc[0]
```

We can use `:` to define a slice of rows we want, such as selecting the second, third,
and fourth rows:

```python
# Select three rows
dataframe.iloc[1:4]
```

We can even use it to get all rows up to a point, such as **all rows up to and including the fourth row:**

```python
# Select three rows
dataframe.iloc[:4]
```

```python
# second, third rows having all columns
dataframe.iloc[1:3, :]
```

```python
# second, third rows with only Name and Age columns
dataframe.iloc[1:3, :2]
```

### Selecting Rows Based on Conditionals

```python
# Show rows where column 'sex' is 'female'
dataframe[dataframe['Sex'] == 'Female']
```

Multiple conditions are easy as well.

```python
# Filter rows
dataframe[(dataframe['Sex'] == 'Female') & (dataframe['Age'] >= 27)]
```

### Replacing Values

pandas’ replace is an easy way to find and replace values. For example, we can
replace any instance of "female" in the Sex column with "Woman":

```python
# Replace values, show two rows
dataframe['Sex'].replace("Female", "Woman")
```

We can also replace multiple values at the same time:

```python
# Replace "female" and "male with "Woman" and "Man"
dataframe['Sex'].replace(["Female", "Male"], ["Woman", "Man"])
```

### Finding the Minimum, Maximum, Sum, Average, and Count

```python
print('Maximum:', dataframe['Age'].max())
print('Minimum:', dataframe['Age'].min())
print('Mean:', dataframe['Age'].mean())
print('Sum:', dataframe['Age'].sum())
print('Count:', dataframe['Age'].count())
```

```python
# Show counts
dataframe.count()         # count non null values
```

###  Deleting a Column

The best way to delete a column is to use drop with the parameter axis=1 (i.e., the
column axis):

```python
# Delete column
dataframe.drop('Age', axis=1).head(2)
```

You can also use a list of column names as the main argument to drop multiple columns
at once:

```python
# Drop columns
dataframe.drop(['Age', 'Sex'], axis=1).head(2)
```

If a column does not have a name (which can sometimes happen), you can drop it by
its column index using dataframe.columns:

```python
# Drop column
dataframe.drop(dataframe.columns[1], axis=1).head(2)
```

### Deleting a Row

### Concatenating DataFrames

Use concat with axis=0 to concatenate along the row axis:

```python
# Load library
import pandas as pd
# Create DataFrame
data_a = {'id': ['1', '2', '3'],
'first': ['Alex', 'Amy', 'Allen'],
'last': ['Anderson', 'Ackerman', 'Ali']}
dataframe_a = pd.DataFrame(data_a, columns = ['id', 'first', 'last'])
# Create DataFrame
data_b = {'id': ['4', '5', '6'],
'first': ['Billy', 'Brian', 'Bran'],
'last': ['Bonder', 'Black', 'Balwner']}
dataframe_b = pd.DataFrame(data_b, columns = ['id', 'first', 'last'])
# Concatenate DataFrames by rows
pd.concat([dataframe_a, dataframe_b], axis=0)
```

You can use axis=1 to concatenate along the column axis:

```python
# Concatenate DataFrames by columns
pd.concat([dataframe_a, dataframe_b], axis=1)
```
