```python
def find(parent, node):
    while parent[node] != node:
        node = parent[node]
    return node

def union(parent, rank, a, b):
    root_a = find(parent, a)
    root_b = find(parent, b)
    if rank[root_a] < rank[root_b]:
        parent[root_a] = root_b
    elif rank[root_a] > rank[root_b]:
        parent[root_b] = root_a
    else:
        parent[root_b] = root_a
        rank[root_a] += 1

def bubble_sort_edges(edges):
    n = len(edges)
    for i in range(n - 1):
        for j in range(n - i - 1):
            if edges[j][2] > edges[j+1][2]:
                edges[j], edges[j+1] = edges[j+1], edges[j]
    return edges

def kruskal(num_nodes, edges):
    
    edges = bubble_sort_edges(edges)  # correct function call

    parent = []
    rank = []
    for i in range(num_nodes):
        parent.append(i)
        rank.append(0)

    mst = []
    mst_weight = 0
    edge_count = 0

    print("\nEdge       Weight   Status")
    print("-" * 35)

    for edge in edges:
        node1  = edge[0]
        node2  = edge[1]
        weight = edge[2]

        root1 = find(parent, node1)
        root2 = find(parent, node2)

        if root1 != root2:
            union(parent, rank, node1, node2)
            mst.append(edge)
            mst_weight += weight
            edge_count += 1
            print(f"  {node1} -- {node2}      {weight}        Added ")
        else:
            print(f"  {node1} -- {node2}      {weight}        Skipped (cycle)")

        if edge_count == num_nodes - 1:
            break

    print(f"\nMST Total Weight = {mst_weight}")
    return mst, mst_weight


# Test
edges = [
    [0, 1, 4],
    [0, 2, 2],
    [0, 3, 6],
    [1, 3, 3],
    [2, 3, 1],
]
num_nodes = 4

print("Nodes: A=0, B=1, C=2, D=3")
mst, weight = kruskal(num_nodes, edges)
```

# **PRIM'S ALGORITHM**

```python
def prim(num_nodes, graph):
    """
    num_nodes : total nodes
    graph     : adjacency matrix → graph[i][j] = weight (0 means no edge)
    """

    visited    = [False] * num_nodes  # track which nodes are in MST
    min_weight = [99999]  * num_nodes  # minimum edge weight to reach each node
    parent     = [-1]    * num_nodes  # stores MST structure

    # Step 1: Start from node 0
    min_weight[0] = 0

    mst_weight = 0

    print("\n--- Prim's MST ---")
    print("Edge       Weight")
    print("-" * 25)

    for _ in range(num_nodes):

        # Step 2: Find unvisited node with minimum weight (manual min search)
        u = -1
        for v in range(num_nodes):
            if not visited[v]:
                if u == -1 or min_weight[v] < min_weight[u]:
                    u = v

        # Step 3: Mark it visited
        visited[u] = True
        mst_weight += min_weight[u]

        if parent[u] != -1:     # skip starting node
            print(f"  {parent[u]} -- {u}      {min_weight[u]}")

        # Step 4: Update weights of neighboring nodes
        for v in range(num_nodes):
            if graph[u][v] != 0 and not visited[v]:    # there is an edge
                if graph[u][v] < min_weight[v]:         # found cheaper path
                    min_weight[v] = graph[u][v]
                    parent[v]     = u

    print(f"\n MST Total Weight = {mst_weight}")
    return mst_weight


# --- Test ---
# Adjacency Matrix (0 = no edge)
# Nodes: A=0, B=1, C=2, D=3
graph = [
    [0, 4, 2, 6],   # A connects to B(4), C(2), D(6)
    [4, 0, 0, 3],   # B connects to A(4), D(3)
    [2, 0, 0, 1],   # C connects to A(2), D(1)
    [6, 3, 1, 0],   # D connects to A(6), B(3), C(1)
]

print("Nodes: A=0, B=1, C=2, D=3")
prim(4, graph)
```
