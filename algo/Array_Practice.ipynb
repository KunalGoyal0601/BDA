# **2. Write a program to find the largest and smallest elements in a list without using buit in functions**

```python
def max_min_array(arr):
    if length_array(arr) == 0:
        return None, None
    small = arr[0]
    large = arr[0]
    for i in range(1, length_array(arr)):
        if arr[i] < small:
            small = arr[i]
        if arr[i] > large:
            large = arr[i]
    return small, large
```

```python
Array = [34, -50, 23, 0, 89, -1, 67]
minimum, maximum = max_min_array(Array)
print("Minimum:", minimum)
print("Maximum:", maximum)
```

# **3. Write a program to find the second largest elements in a list**

```python
def sort_array(arr):
    n = length_array(arr)
    for i in range(n):
        for j in range(0,n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr
```

```python
Array = [34, -50, 23, 0, 89, -1, 67]
Sorted_array = sort_array(Array)
print(f"Sorted Array:,{Sorted_array}\nSecond largedt element-> {Sorted_array[-2]}")
```

# **6. Create a NumPy array of 10 random integers and find: Mean, Median, Standard deviation**

```python
arr = np.random.randint(1,100, size=10)
print("Random Array:", arr)
```

**Mean**

```python
#finding mean
def find_mean(arr):
    total = 0
    for num in arr:
        total += num
        mean = total / length_array(arr)
    return total, mean
```

```python
total_,mean_ = find_mean(arr)
print("Array for mean and total calculation:", arr)
print(f"Total is -> {total_} \nMean is -> {mean_}")
```

**Median**

```python
#find median
def find_median(arr):
    Sorted_array = sort_array(arr)
    n =length_array(Sorted_array)
    if n%2 == 0:
        median = (Sorted_array[n//2 + 1] + Sorted_array[n//2]) / 2
    else:
        median = Sorted_array[n//2]
    return median
```

```python
median_ = find_median(arr)
print("Array for median calculation:", arr)
print("Median is ->", median_)
```

**Standard deviation**

```python
#find standard deviation
def find_standard_deviation(arr):
    total, mean = find_mean(arr)
    variance_sum = 0
    for num in arr:
        variance_sum += (num - mean) ** 2
    variance = variance_sum / length_array(arr)
    standard_deviation = variance ** 0.5
    return standard_deviation
```

```python
standard_deviation_ = find_standard_deviation(arr)
print("Array for standard deviation calculation:", arr)
print("Standard Deviation is ->", standard_deviation_)
```

# **12. Reverse the elements of a list without using reverse().**

```python
def reverse_list(arr):
    start = 0
    end = length_array(arr) - 1
    while start < end:
        arr[start],arr[end] = arr[end], arr[start]
        start +=1
        end -=1
    return arr
```

```python
arr = [34, 89, 23, 0, 89, 67, 67, 23, 23, 0]
reverased_arr = reverse_list(arr)
print("Reversed array:", reverased_arr)
```

# **14. Write a program to add two matrices.**

# **15. Find the determinant of a matrix.**

```python
def determinant(matrix):
    # Check if the matrix is empty
    if length_array(matrix) == 0:
        return None
    # Check if the matrix is square
    if length_array(matrix) != length_array(matrix[0]):
        return None
    # Base case for 1x1 matrix
    if length_array(matrix) == 1:
        return matrix[0][0]
    # Base case for 2x2 matrix
    if length_array(matrix) == 2:
        return (matrix[0][0] * matrix[1][1]) - (matrix[0][1] * matrix[1][0])
    # Recursive case: cofactor expansion along the first row
    det = 0
    for col in range(length_array(matrix)):
        # Build the minor matrix (remove row 0 and current column)
        minor = [[matrix[row][c] for c in range(length_array(matrix)) if c != col]
            for row in range(1, length_array(matrix))]
        # Alternating sign: + - + - ...
        sign = (-1) ** col
        det += sign * matrix[0][col] * determinant(minor)

    return det
```

```python
matrix2x2 =  [
        [3, 8],
        [4, 6]
    ]

matrix3x3 =  [
        [1, 2, 3],
        [4, 5, 6],
        [7, 8, 9]
    ]
     
matrix4x4 =       [
        [1, 3, 5, 9],
        [1, 3, 1, 7],
        [4, 3, 9, 7],
        [5, 2, 0, 9]
    ]

print("Determinant of 2x2 matrix is:", determinant(matrix2x2))
print("Determinant of 3x3 matrix is:", determinant(matrix3x3))
print("Determinant of 4x4 matrix is:", determinant(matrix4x4))
```

# **16. Check whether a matrix is an identity matrix.**

```python
def identity_matrix_chack(matrix):
    if length_array(matrix) == 0:
        return False
    for i in range(length_array(matrix)):
        for j in range(length_array(matrix[0])):
            if i == j:
                if matrix[i][j] != 1:
                    return False
            else:
                if matrix[i][j] != 0:
                    return False
    return True
```

```python
matrix = [[1,0,0],[0,1,0],[0,0,1]]
is_identity = identity_matrix_chack(matrix)
print("Is the matrix an identity matrix?", is_identity)
```

# **17. WAP to perform the union and intersection of arrays**

**Union**

```python
def union(arr1,arr2):
    union_set = []
    for num in arr1:
        if num not in union_set:
            union_set=union_set + [num]
    for num in arr2:
        if num not in union_set:
            union_set=union_set + [num]
    return union_set
```

```python
arr1 = [1,2,3,4,5]
arr2 = [4,5,6,7,8]
union_result = union(arr1,arr2)
print("Union of the two arrays:", union_result)
```

**Intersection**

```python
def intersection(arr1,arr2):
    intersection_set = []
    for num in arr1:
        if num in arr2 and num not in intersection_set:
            intersection_set = intersection_set + [num]
    return intersection_set
```

```python
arr1 = [1,2,3,4,5]
arr2 = [4,5,6,7,8]
intersection_result = intersection(arr1,arr2)
print("Intersection of the two arrays:", intersection_result)
```

# **20. Write a program (WAP) to perform Inversion Count: how far (or close) the array is from being sorted**

```python
def inversion_count(arr):
    count = 0
    for i in range(length_array(arr)):
        for j in range(i+1, length_array(arr)):
            if arr[i] > arr[j]:
                count +=1
    return count
```

```python
arr = [34, -50, 23, 0, 89, -1, 67]
inversions = inversion_count(arr)
print("Number of inversions in the array:", inversions)
```

# **21. Write a program (WAP) to perform Two Sum-Find: whether there is a pair of elements in the array whose sum is equal to target.**

```python
def two_sum(arr, target):
    for i in range(length_array(arr)):
        for j in range(i+1, length_array(arr)):
            if arr[i]+arr[j] == target:
                return (arr[i],arr[j])
```

```python
arr = [34, -50, 23, 0, 89, -1, 67]
target_var = 90
result = two_sum(arr, target_var)
print("Indices of elements that sum to", target_var, "are:", result)
```
