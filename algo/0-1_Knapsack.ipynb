# **0/1 Knapsack**
---

```python
def knapsack_01(capacity, items):
    """
    capacity : max weight knapsack can hold
    items    : list of [value, weight]
    """
    n = len(items)

    # Step 1: Build DP table of size (n+1) x (capacity+1)
    # dp[i][w] = best value with i items, capacity w
    dp = []
    for i in range(n + 1):
        row = []
        for w in range(capacity + 1):
            row.append(0)       # fill with 0 initially
        dp.append(row)

    # Step 2: Fill the table row by row
    for i in range(1, n + 1):
        value  = items[i-1][0]
        weight = items[i-1][1]

        for w in range(capacity + 1):

            # Choice 1: Skip this item
            skip = dp[i-1][w]

            # Choice 2: Take this item (only if it fits)
            take = 0
            if weight <= w:
                take = value + dp[i-1][w - weight]

            # Pick the better choice
            if take > skip:
                dp[i][w] = take
            else:
                dp[i][w] = skip

    # Step 3: Print DP Table
    print("\n--- DP Table ---")
    print("     ", end="")
    for w in range(capacity + 1):
        print(f"{w:<4}", end="")
    print()
    print("-" * (capacity + 2) * 4)

    for i in range(n + 1):
        print(f"i={i} | ", end="")
        for w in range(capacity + 1):
            print(f"{dp[i][w]:<4}", end="")
        print()


    # Step 4: Find which items were selected
    # (Trace back through the table
    selected = []
    w = capacity
    for i in range(n, 0, -1):
        if dp[i][w] != dp[i-1][w]:     # value changed = item was taken
            selected.append(i)
            w -= items[i-1][1]          # reduce capacity

    # Step 5: Print Results
    print("\n--- Selected Items ---")
    print("Item   Value   Weight")
    print("-" * 25)
    for i in selected:
        print(f"  {i}      {items[i-1][0]}      {items[i-1][1]}")

    print(f"\nMaximum Value = {dp[n][capacity]}")
    return dp[n][capacity]


# --- Test ---
items = [
    [60,  10],   # item 1: value=60,  weight=10
    [100, 20],   # item 2: value=100, weight=20
    [120, 30],   # item 3: value=120, weight=30
]
capacity = 50

print(f"Capacity : {capacity}")
print(f"Items    : {len(items)}")
knapsack_01(capacity, items)   
```
