# 🔍 Algorithm 1: Linear Search

---

## 1️⃣ Introduction

### What is Linear Search?
Linear Search is the **simplest searching algorithm**. It checks **every element one by one** from the beginning until it finds the target or reaches the end.

### Real-Life Analogy 🏠
Imagine you lost your phone at home. You search **every room one by one** — bedroom, kitchen, bathroom, living room — until you find it or run out of rooms.

That is exactly Linear Search!

### Where Is It Used?
- Searching in **unsorted lists**
- When the data is **small**
- When elements are **not in any order**
- Searching in a **linked list**

---

## 2️⃣ Problem Understanding

**Problem:** Given a list of numbers and a target, find the **position (index)** of the target in the list.

**Example:**
```
Array  = [4, 2, 9, 7, 5, 1]
Target = 7
Answer = Index 3
```

**Why not just use a shortcut?**  
There is no shortcut for an **unsorted list**. We MUST check every element.

---

## 3️⃣ Core Intuition

The main idea is simple:

> **"Check each element. If it matches the target → return its index. If we finish the list without finding it → return -1."**

- We start from **index 0** (the first element).
- We move one step at a time → **index 1, 2, 3...**
- If we find the target → **stop and return the index**
- If we reach the end without finding → **return -1** (not found)

---

## 4️⃣ Dry Run Example

**Array = [4, 2, 9, 7, 5, 1], Target = 7**

| Step | Index | Element Checked | Match? | Action |
|------|-------|----------------|--------|--------|
| 1    | 0     | 4              | ❌ No  | Move right |
| 2    | 1     | 2              | ❌ No  | Move right |
| 3    | 2     | 9              | ❌ No  | Move right |
| 4    | 3     | **7**          | ✅ Yes | **Return index 3** |

**Total comparisons = 4**

---

**Array = [4, 2, 9, 7, 5, 1], Target = 8** (element NOT in array)

| Step | Index | Element Checked | Match? | Action |
|------|-------|----------------|--------|--------|
| 1    | 0     | 4              | ❌ No  | Move right |
| 2    | 1     | 2              | ❌ No  | Move right |
| 3    | 2     | 9              | ❌ No  | Move right |
| 4    | 3     | 7              | ❌ No  | Move right |
| 5    | 4     | 5              | ❌ No  | Move right |
| 6    | 5     | 1              | ❌ No  | Move right |
| End  | —     | List exhausted | —      | **Return -1** |

---

## 5️⃣ Step-by-Step Algorithm (Plain English)

```
Step 1: Start at index 0 (the first position)
Step 2: Compare the current element with the target
Step 3: If they match → return the current index (FOUND!)
Step 4: If they don't match → move to the next index
Step 5: Repeat Steps 2–4 until:
          - The element is found, OR
          - We reach the end of the list
Step 6: If we finish the list without finding → return -1 (NOT FOUND)
```


````python
# ============================================================
# LINEAR SEARCH — Full Python Code (From Scratch)
# ============================================================

def linear_search(arr, target):
    """
    Searches for 'target' in list 'arr'.
    Returns the index if found, else returns -1.
    """
    
    # Step 1: Loop through every index in the array
    for i in range(len(arr)):
        
        # Step 2: Check if current element equals target
        if arr[i] == target:
            return i          # Found! Return the index.
    
    # Step 3: If loop finishes without finding, return -1
    return -1


# ─── Test Cases ───────────────────────────────────────────

arr = [4, 2, 9, 7, 5, 1]

# Test 1: Target exists in array
result = linear_search(arr, 7)
print(f"Search for 7  → Index: {result}")   # Expected: 3

# Test 2: Target does NOT exist
result = linear_search(arr, 8)
print(f"Search for 8  → Index: {result}")   # Expected: -1

# Test 3: Target is the first element (Best Case)
result = linear_search(arr, 4)
print(f"Search for 4  → Index: {result}")   # Expected: 0

# Test 4: Target is the last element (Worst Case)
result = linear_search(arr, 1)
print(f"Search for 1  → Index: {result}")   # Expected: 5
````

## 7️⃣ Line-by-Line Code Explanation

| Line | Code | What It Does |
|------|------|-------------|
| `for i in range(len(arr))` | Loop variable `i` goes 0, 1, 2, ... len-1 | Visits every index |
| `if arr[i] == target` | Compare element at index `i` with our target | Check if found |
| `return i` | Return the index immediately | Stops loop early if found |
| `return -1` | After loop ends with no match | Signals "not found" |

---

## 9️⃣ Complexity Analysis

| Case | Situation | Comparisons | Time Complexity |
|------|-----------|-------------|-----------------|
| **Best Case** | Target is the **first element** | 1 comparison | **O(1)** |
| **Average Case** | Target is somewhere in the **middle** | n/2 comparisons | **O(n)** |
| **Worst Case** | Target is **last** OR **not found** | n comparisons | **O(n)** |
| **Space** | No extra space used | — | **O(1)** |

### Why O(n)?
In the worst case, we look at **every element once**. If the array has `n` elements, we do at most `n` comparisons. So the time grows **linearly** with `n` → **O(n)**.

# 🔍 Algorithm 2: Binary Search

---

## 1️⃣ Introduction

### What is Binary Search?
Binary Search is a **much faster** searching algorithm — but it only works on a **sorted array**. Instead of checking every element, it **cuts the search space in HALF** every time.

### Real-Life Analogy 📖
Think of finding a word in a **dictionary**:
1. Open to the **middle page**.
2. If your word comes **before** the middle → search the **left half**.
3. If your word comes **after** the middle → search the **right half**.
4. Repeat until you find it.

You'd **never** read every word from A to Z. You **halve** the search space every time!

### Why Must the Array Be Sorted?
If the array is unsorted, we can't know **which half** our target is in. The "left or right" decision only works when elements are in order.

---

## 2️⃣ Problem Understanding

**Problem:** Given a **sorted** array and a target, find the index of the target.

```
Array  = [1, 3, 5, 7, 9, 11, 13, 15]
Target = 7
Answer = Index 3
```

**Why not Linear Search here?**
Linear Search would check: 1, 3, 5, 7 → 4 comparisons.
Binary Search finds it in **2 comparisons**. On a million elements, Binary Search wins by a HUGE margin!

---

## 3️⃣ Core Intuition

> **"Eliminate half the remaining elements at every step."**

We maintain three pointers:
- `left` → Start of the search range
- `right` → End of the search range
- `mid` → Middle of the current range

Each comparison tells us: **"Is the target in the left half or right half?"**

---

## 4️⃣ Dry Run Example

**Array = [1, 3, 5, 7, 9, 11, 13, 15], Target = 7**

```
Indices:  0   1   2   3   4    5    6    7
Array  = [1,  3,  5,  7,  9,  11,  13,  15]
```

**Step 1:**
```
left=0, right=7
mid = 0 + (7-0)//2 = 3
arr[3] = 7 == target ✅  → FOUND at index 3!
```

**Another example — Target = 11:**

**Step 1:**
```
left=0, right=7,  mid=3,  arr[3]=7  < 11  → search RIGHT half
left becomes mid+1 = 4
```
**Step 2:**
```
left=4, right=7,  mid=5,  arr[5]=11 == 11 ✅ → FOUND at index 5!
```

**Another example — Target = 6 (not in array):**

**Step 1:** `left=0, right=7, mid=3, arr[3]=7 > 6` → search LEFT, `right = 2`  
**Step 2:** `left=0, right=2, mid=1, arr[1]=3 < 6` → search RIGHT, `left = 2`  
**Step 3:** `left=2, right=2, mid=2, arr[2]=5 < 6` → search RIGHT, `left = 3`  
**Step 4:** `left=3 > right=2` → **STOP. Return -1 (not found).**

---

## 5️⃣ Step-by-Step Algorithm

```
Step 1: Set left = 0, right = len(array) - 1
Step 2: While left <= right:
    Step 3: Compute mid = left + (right - left) // 2
    Step 4: If arr[mid] == target → return mid (FOUND!)
    Step 5: If arr[mid] < target  → left = mid + 1  (search right half)
    Step 6: If arr[mid] > target  → right = mid - 1 (search left half)
Step 7: Return -1 (NOT FOUND)
```

### ⚠️ Why `mid = left + (right - left) // 2` and NOT `(left + right) // 2`?

Using `(left + right) // 2` can cause an **integer overflow** in languages like Java/C++ when left and right are very large.
`left + (right - left) // 2` is always safe. Good habit to form!


````python
# ============================================================
# BINARY SEARCH — Iterative Approach
# ============================================================

def binary_search_iterative(arr, target):
    """
    Binary Search using a WHILE loop.
    Works on a SORTED array.
    Returns index if found, else -1.
    """
    
    left = 0              # Start of search range
    right = len(arr) - 1  # End of search range
    
    while left <= right:  # Keep searching while range is valid
        
        # Calculate mid safely (avoids integer overflow in other languages)
        mid = left + (right - left) // 2
        
        if arr[mid] == target:       # 🎯 Found the target!
            return mid
        
        elif arr[mid] < target:      # Target is in the RIGHT half
            left = mid + 1           # Discard left half
        
        else:                        # Target is in the LEFT half
            right = mid - 1          # Discard right half
    
    return -1   # Target not found


# ─── Test Cases ───────────────────────────────────────────

arr = [1, 3, 5, 7, 9, 11, 13, 15]

print("Array:", arr)
print()

result = binary_search_iterative(arr, 7)
print(f"Search for  7 → Index: {result}")   # Expected: 3

result = binary_search_iterative(arr, 11)
print(f"Search for 11 → Index: {result}")   # Expected: 5

result = binary_search_iterative(arr, 1)
print(f"Search for  1 → Index: {result}")   # Expected: 0 (Best case)

result = binary_search_iterative(arr, 15)
print(f"Search for 15 → Index: {result}")   # Expected: 7

result = binary_search_iterative(arr, 6)
print(f"Search for  6 → Index: {result}")   # Expected: -1 (not found)
````

````python
# ============================================================
# BINARY SEARCH — Recursive Approach
# ============================================================

def binary_search_recursive(arr, target, left, right):
    """
    Binary Search using RECURSION.
    Call it as: binary_search_recursive(arr, target, 0, len(arr)-1)
    """
    
    # BASE CASE: If left > right, search space is empty → not found
    if left > right:
        return -1
    
    # Calculate mid
    mid = left + (right - left) // 2
    
    if arr[mid] == target:            # 🎯 Found!
        return mid
    
    elif arr[mid] < target:           # Search RIGHT half
        return binary_search_recursive(arr, target, mid + 1, right)
    
    else:                             # Search LEFT half
        return binary_search_recursive(arr, target, left, mid - 1)


# ─── Test Cases ───────────────────────────────────────────

arr = [1, 3, 5, 7, 9, 11, 13, 15]

print("Array:", arr)
print()

result = binary_search_recursive(arr, 7, 0, len(arr) - 1)
print(f"Recursive search for  7 → Index: {result}")   # Expected: 3

result = binary_search_recursive(arr, 11, 0, len(arr) - 1)
print(f"Recursive search for 11 → Index: {result}")   # Expected: 5

result = binary_search_recursive(arr, 6, 0, len(arr) - 1)
print(f"Recursive search for  6 → Index: {result}")   # Expected: -1
````

## 7️⃣ Line-by-Line Code Explanation

### Iterative Version:
| Line | Explanation |
|------|-------------|
| `left = 0` | Start searching from the beginning |
| `right = len(arr) - 1` | Search ends at the last index |
| `while left <= right` | Continue as long as there's a valid search range |
| `mid = left + (right-left)//2` | Find the middle safely |
| `arr[mid] == target` | Check if we found the answer |
| `left = mid + 1` | Throw away left half; search right |
| `right = mid - 1` | Throw away right half; search left |
| `return -1` | Loop ended: target not found |

### Recursive Version:
| Line | Explanation |
|------|-------------|
| `if left > right: return -1` | Base case: search space exhausted |
| `return binary_search_recursive(..., mid+1, right)` | Recursive call on right half |
| `return binary_search_recursive(..., left, mid-1)` | Recursive call on left half |

## 9️⃣ Complexity Analysis

| Case | Situation | Time Complexity |
|------|-----------|-----------------|
| **Best Case** | Target is at the **mid** of first check | O(1) |
| **Average Case** | Target found after a few halvings | O(log n) |
| **Worst Case** | Target not found (or at boundary) | O(log n) |
| **Space (Iterative)** | Only a few variables | O(1) |
| **Space (Recursive)** | Recursion call stack grows | O(log n) |

### Why O(log n)?
Each step, we **halve** the search space:
- Start: n elements
- After 1 step: n/2
- After 2 steps: n/4
- After k steps: n/2^k

We stop when n/2^k = 1 → k = log₂(n)

**Example:** Array of 1,000,000 elements → max **20 comparisons!** (log₂(1M) ≈ 20)
Linear Search would need up to **1,000,000** comparisons!

# 🫧 Algorithm 3: Bubble Sort

---

## 1️⃣ Introduction

### What is Bubble Sort?
Bubble Sort repeatedly **compares two adjacent (neighboring) elements** and **swaps them if they are in the wrong order**. The largest elements "bubble up" to the end, like bubbles rising in water.

### Real-Life Analogy 🫧
Imagine students standing in a line, and a teacher checks pairs of students from left to right. If a taller student is to the LEFT of a shorter one, they swap places. After one full pass, the **tallest student is at the rightmost end**. Repeat for the remaining students.

### Where Is It Used?
- Educational purposes (easiest sorting algorithm to understand)
- When array is **nearly sorted** (optimized version is very fast)
- Rarely used in production (other algorithms are faster)

---

## 2️⃣ Problem Understanding

**Problem:** Sort an unsorted array in ascending order.
```
Input:  [64, 25, 12, 22, 11]
Output: [11, 12, 22, 25, 64]
```

---

## 3️⃣ Core Intuition

> **"In each pass, push the largest unsorted element to its correct position at the end."**

- After Pass 1 → Largest element is at the END ✅
- After Pass 2 → 2nd largest is at second-last position ✅
- After Pass k → k largest elements are correctly placed ✅

---

## 4️⃣ Dry Run Example

**Array = [5, 3, 8, 4, 2]**

### Pass 1 (i=0): Compare adjacent pairs
```
[5, 3, 8, 4, 2]
 ↑  ↑
 5 > 3? YES → Swap → [3, 5, 8, 4, 2]

[3, 5, 8, 4, 2]
    ↑  ↑
    5 > 8? NO → No swap

[3, 5, 8, 4, 2]
       ↑  ↑
       8 > 4? YES → Swap → [3, 5, 4, 8, 2]

[3, 5, 4, 8, 2]
          ↑  ↑
          8 > 2? YES → Swap → [3, 5, 4, 2, 8]
                                              ✅ 8 is now at correct position!
```
### Pass 2 (i=1):
```
[3, 5, 4, 2, 8] → [3, 4, 2, 5, 8]   ← 5 bubbles to correct spot
```
### Pass 3 (i=2):
```
[3, 4, 2, 5, 8] → [3, 2, 4, 5, 8]   ← 4 is placed
```
### Pass 4 (i=3):
```
[3, 2, 4, 5, 8] → [2, 3, 4, 5, 8]   ← Fully sorted! ✅
```

---

## 5️⃣ Step-by-Step Algorithm

```
Step 1: Repeat n-1 times (outer loop, pass number i)
Step 2:   For each adjacent pair from index 0 to (n-2-i):
Step 3:     If arr[j] > arr[j+1]:
Step 4:       Swap arr[j] and arr[j+1]
Step 5:     (Optional) If no swap happened in a full pass → array is sorted → STOP early
```


````python
# ============================================================
# BUBBLE SORT — Full Python Code (From Scratch)
# ============================================================

def bubble_sort(arr):
    """
    Sorts the array in ascending order using Bubble Sort.
    Includes the 'swapped' optimization for early exit.
    """
    
    n = len(arr)
    
    # Outer loop: controls how many passes we do
    for i in range(n - 1):
        
        swapped = False   # Optimization: track if any swap happened
        
        # Inner loop: compare adjacent pairs
        # After each pass i, the last i elements are already sorted
        # so we only go up to (n - 1 - i)
        for j in range(n - 1 - i):
            
            if arr[j] > arr[j + 1]:          # Wrong order?
                arr[j], arr[j + 1] = arr[j + 1], arr[j]   # Swap!
                swapped = True
        
        # If no swaps happened, the array is already sorted → stop early
        if not swapped:
            print(f"  ✅ Early exit at pass {i+1} — array already sorted!")
            break
    
    return arr


# ─── Step-by-step visualization version ───────────────────

def bubble_sort_verbose(arr):
    """Same as bubble_sort but prints every step."""
    arr = arr[:]   # Work on a copy so original is unchanged
    n = len(arr)
    
    print(f"Initial array: {arr}\n")
    
    for i in range(n - 1):
        print(f"--- Pass {i + 1} ---")
        swapped = False
        
        for j in range(n - 1 - i):
            print(f"  Compare arr[{j}]={arr[j]} and arr[{j+1}]={arr[j+1]}  ", end="")
            
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
                print(f"→ SWAP → {arr}")
            else:
                print(f"→ No swap")
        
        print(f"  After Pass {i+1}: {arr}\n")
        
        if not swapped:
            print("✅ No swaps in this pass → array is sorted!")
            break
    
    return arr


# ─── Test ─────────────────────────────────────────────────

print("=" * 50)
print("BUBBLE SORT VERBOSE DEMO")
print("=" * 50)
result = bubble_sort_verbose([5, 3, 8, 4, 2])
print(f"\nFinal sorted array: {result}")

print("\n" + "=" * 50)

# Test with already-sorted array (shows early exit)
print("\nTesting EARLY EXIT optimization:")
bubble_sort_verbose([1, 2, 3, 4, 5])
````

## 7️⃣ Line-by-Line Code Explanation

| Line | Code | Explanation |
|------|------|-------------|
| `for i in range(n-1)` | Outer loop | We need at most n-1 passes |
| `swapped = False` | Reset flag each pass | Track if any swap occurred |
| `for j in range(n-1-i)` | Inner loop | Skip already-sorted tail |
| `if arr[j] > arr[j+1]` | Compare neighbors | Is the pair in wrong order? |
| `arr[j], arr[j+1] = arr[j+1], arr[j]` | Python swap | Exchange the two elements |
| `swapped = True` | Mark that a swap happened | Used for early exit |
| `if not swapped: break` | Early exit | If no swap → already sorted! |

### 🔑 Key Insight: Why `n-1-i` in the inner loop?
After each pass `i`, the **last `i` elements are already in their correct positions**. No need to check them again. This makes Bubble Sort slightly faster.

---

## 8️⃣ Visualization

```
Pass 1:  [5, 3, 8, 4, 2]  →  [3, 5, 4, 2, 8]   ← 8 placed ✅
Pass 2:  [3, 5, 4, 2, 8]  →  [3, 4, 2, 5, 8]   ← 5 placed ✅
Pass 3:  [3, 4, 2, 5, 8]  →  [3, 2, 4, 5, 8]   ← 4 placed ✅
Pass 4:  [3, 2, 4, 5, 8]  →  [2, 3, 4, 5, 8]   ← Fully sorted ✅
```

## 9️⃣ Complexity Analysis

| Case | Situation | Time Complexity |
|------|-----------|-----------------|
| **Best Case** | Array already sorted (with optimization) | **O(n)** |
| **Average Case** | Random order | **O(n²)** |
| **Worst Case** | Array in reverse order | **O(n²)** |
| **Space** | No extra array needed | **O(1)** |

### Why O(n²)?
- Outer loop runs n-1 times
- Inner loop runs n-1, n-2, n-3... 1 times
- Total = (n-1) + (n-2) + ... + 1 = n(n-1)/2 ≈ **n²**

---

## 🚨 10. Common Beginner Mistakes

| Mistake | Effect | Fix |
|---------|--------|-----|
| `range(n)` in inner loop | **Index out of bounds** at `arr[j+1]` | Use `range(n-1-i)` |
| Forgetting `-i` in inner loop | Checks already-sorted elements again | Add `-i` |
| `arr[j] >= arr[j+1]` | Unnecessary swaps of equal elements | Use `>` not `>=` |
| Not copying the array | Modifies the original accidentally | Use `arr[:]` for a copy |

---

**🟢 Easy:** Sort in **descending** order using Bubble Sort.  
**🟡 Medium:** Count the total number of swaps made.  
**🔴 Hard:** Sort a list of strings alphabetically using Bubble Sort.

---

| Question | Answer |
|----------|--------|
| Is Bubble Sort stable? | **Yes** — equal elements maintain relative order |
| Is it in-place? | **Yes** — O(1) extra space |
| Best use case? | Nearly-sorted arrays (with optimization → O(n)) |
| Why is it rarely used? | O(n²) is too slow for large datasets |

# 🎯 Algorithm 4: Selection Sort

---

## 1️⃣ Introduction

### What is Selection Sort?
Selection Sort divides the array into two parts:
1. **Sorted part** (left side, starts empty)
2. **Unsorted part** (right side, starts as the whole array)

In each step, it **finds the MINIMUM element** from the unsorted part and **places it at the start** of the unsorted part (i.e., swaps it to the correct position).

### Real-Life Analogy 🃏
Imagine picking the **smallest card** from a shuffled deck, putting it face down on the table, then picking the **next smallest**, and so on. You SELECT the minimum each time.

---

## 2️⃣ Problem Understanding

```
Input:  [64, 25, 12, 22, 11]
Output: [11, 12, 22, 25, 64]
```

---

## 3️⃣ Core Intuition

> **"In each pass, find the minimum from the unsorted part and place it in its correct position."**

---

## 4️⃣ Dry Run Example

**Array = [64, 25, 12, 22, 11]**

```
Pass 1: Find min from [64, 25, 12, 22, 11] → min=11 at idx 4
        Swap arr[0] with arr[4] → [11, 25, 12, 22, 64]
                                    ✅

Pass 2: Find min from [25, 12, 22, 64] → min=12 at idx 2
        Swap arr[1] with arr[2] → [11, 12, 25, 22, 64]
                                        ✅

Pass 3: Find min from [25, 22, 64] → min=22 at idx 3
        Swap arr[2] with arr[3] → [11, 12, 22, 25, 64]
                                            ✅

Pass 4: Find min from [25, 64] → min=25 at idx 3 (already in place)
        No actual movement needed → [11, 12, 22, 25, 64]
                                                ✅

Done! ✅ [11, 12, 22, 25, 64]
```

---

## 5️⃣ Step-by-Step Algorithm

```
Step 1: For i from 0 to n-2 (each position to fill):
Step 2:   Assume arr[i] is the minimum (min_idx = i)
Step 3:   For j from i+1 to n-1:
Step 4:     If arr[j] < arr[min_idx]:
Step 5:       Update min_idx = j
Step 6:   Swap arr[i] with arr[min_idx]
```


````python
# ============================================================
# SELECTION SORT — Full Python Code (From Scratch)
# ============================================================

def selection_sort(arr):
    """
    Sorts array in ascending order using Selection Sort.
    In each pass, finds the minimum and places it at position i.
    """
    
    n = len(arr)
    
    # Outer loop: position we're filling (from 0 to n-2)
    for i in range(n - 1):
        
        # Assume the current position has the minimum
        min_idx = i
        
        # Search for the actual minimum in the unsorted part
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j          # Found a smaller element!
        
        # Swap the minimum with the element at position i
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    
    return arr


# ─── Verbose version ──────────────────────────────────────

def selection_sort_verbose(arr):
    arr = arr[:]
    n = len(arr)
    
    print(f"Initial: {arr}\n")
    
    for i in range(n - 1):
        min_idx = i
        
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        
        print(f"Pass {i+1}: min={arr[min_idx]} at index {min_idx}", end="")
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
        print(f" → Swap → {arr}  (sorted portion: {arr[:i+1]})")
    
    return arr


# ─── Test ─────────────────────────────────────────────────

print("=" * 50)
result = selection_sort_verbose([64, 25, 12, 22, 11])
print(f"\nFinal sorted array: {result}")

print()
# Verify correctness
test = [3, 1, 4, 1, 5, 9, 2, 6]
print(f"Input:  {test}")
print(f"Output: {selection_sort(test)}")
````

## 9️⃣ Complexity Analysis

| Case | Time Complexity | Reason |
|------|-----------------|--------|
| **Best Case** | O(n²) | Always scans all unsorted elements |
| **Average Case** | O(n²) | Same nested loop structure |
| **Worst Case** | O(n²) | Same |
| **Space** | O(1) | In-place, no extra array |

> **Note:** Unlike Bubble Sort, Selection Sort does NOT benefit from nearly-sorted input. It always runs n(n-1)/2 comparisons regardless!

**However**, Selection Sort makes at most **n-1 swaps** (one per pass), whereas Bubble Sort can make many more swaps. So Selection Sort can be better when **swapping is expensive**.

---

## 🚨 10. Common Beginner Mistakes

| Mistake | Effect | Fix |
|---------|--------|-----|
| `range(n)` in outer loop | Last element is compared with itself | Use `range(n-1)` |
| `min_idx = 0` always | Finds global min, not local min | Set `min_idx = i` at start of each pass |
| Swapping immediately on finding smaller element | Wrong result | Only swap AFTER finding the TRUE minimum |

---

| Question | Answer |
|----------|--------|
| Is Selection Sort stable? | **No** — swapping can change relative order of equal elements |
| Is it in-place? | **Yes** — O(1) space |
| How many swaps? | At most **n-1** swaps (very efficient in terms of swaps) |
| vs Bubble Sort? | Selection Sort: fewer swaps; Bubble Sort: can exit early |

# 🃏 Algorithm 5: Insertion Sort

---

## 1️⃣ Introduction

### What is Insertion Sort?
Insertion Sort builds the sorted array **one element at a time** by **inserting each new element into its correct position** among the already-sorted elements.

### Real-Life Analogy 🃏
Think of how you sort playing cards in your hand:
- You pick up cards one by one.
- Each new card is **inserted** into its correct position among the cards you're already holding.
- You shift existing cards to make room.

This is **exactly** how Insertion Sort works!

---

## 2️⃣ Problem Understanding

```
Input:  [9, 5, 1, 4, 3]
Output: [1, 3, 4, 5, 9]
```

---

## 3️⃣ Core Intuition

> **"Take each element (starting from index 1) and insert it into its correct place in the sorted left portion."**

- The **left part** grows as sorted elements are placed.
- The **right part** shrinks as we process each element.
- For each element, we **shift** larger elements one position to the right to make room.

---

## 4️⃣ Dry Run Example

**Array = [9, 5, 1, 4, 3]**

```
Initial:  [9 | 5, 1, 4, 3]  ← sorted | unsorted

Step 1: key = 5 (arr[1])
  Compare with 9: 9 > 5 → shift 9 right → [_, 9, 1, 4, 3]
  No more elements left → insert 5 at index 0
  Result: [5, 9 | 1, 4, 3]

Step 2: key = 1 (arr[2])
  Compare with 9: 9 > 1 → shift 9 right → [5, _, 9, 4, 3]
  Compare with 5: 5 > 1 → shift 5 right → [_, 5, 9, 4, 3]
  No more elements → insert 1 at index 0
  Result: [1, 5, 9 | 4, 3]

Step 3: key = 4 (arr[3])
  Compare with 9: 9 > 4 → shift → [1, 5, _, 9, 3]
  Compare with 5: 5 > 4 → shift → [1, _, 5, 9, 3]
  Compare with 1: 1 < 4 → STOP. Insert 4 at index 1.
  Result: [1, 4, 5, 9 | 3]

Step 4: key = 3 (arr[4])
  Shifts: 9 → 5 → 4 → stop at 1
  Insert 3 at index 1
  Result: [1, 3, 4, 5, 9] ✅
```

---

## 5️⃣ Step-by-Step Algorithm

```
Step 1: For i from 1 to n-1 (process each element starting from index 1):
Step 2:   key = arr[i]   (element to be inserted)
Step 3:   j = i - 1     (start comparing from the element before key)
Step 4:   While j >= 0 AND arr[j] > key:
Step 5:     arr[j + 1] = arr[j]   (shift element one position right)
Step 6:     j = j - 1             (move comparison pointer left)
Step 7:   arr[j + 1] = key        (insert key at the correct position)
```


````python
# ============================================================
# INSERTION SORT — Full Python Code (From Scratch)
# ============================================================

def insertion_sort(arr):
    """
    Sorts array in ascending order using Insertion Sort.
    Works by inserting each element into its correct position.
    """
    
    n = len(arr)
    
    # Start from index 1 (index 0 is trivially "sorted" by itself)
    for i in range(1, n):
        
        key = arr[i]    # The element we want to INSERT correctly
        j = i - 1       # Start comparing from the element just before key
        
        # Shift elements that are GREATER than key one position to the RIGHT
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]   # Shift right
            j -= 1                # Move comparison pointer left
        
        # Insert key at its correct position
        arr[j + 1] = key
    
    return arr


# ─── Verbose version ──────────────────────────────────────

def insertion_sort_verbose(arr):
    arr = arr[:]
    n = len(arr)
    
    print(f"Initial: {arr}\n")
    
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        
        print(f"Step {i}: key = {key}")
        
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            print(f"  Shift {arr[j]} right → {arr}")
            j -= 1
        
        arr[j + 1] = key
        print(f"  Insert {key} at index {j+1} → {arr}\n")
    
    return arr


# ─── Test ─────────────────────────────────────────────────

print("=" * 50)
result = insertion_sort_verbose([9, 5, 1, 4, 3])
print(f"Final sorted array: {result}")

print()
# Test best case: already sorted
test_sorted = [1, 2, 3, 4, 5]
print(f"Already sorted: {test_sorted} → {insertion_sort(test_sorted)}")

# Test worst case: reverse sorted
test_reverse = [5, 4, 3, 2, 1]
print(f"Reverse sorted: {test_reverse} → {insertion_sort(test_reverse)}")
````

## 9️⃣ Complexity Analysis

| Case | Situation | Time Complexity |
|------|-----------|-----------------|
| **Best Case** | Array already sorted (inner while loop never runs) | **O(n)** |
| **Average Case** | Random order | **O(n²)** |
| **Worst Case** | Array in reverse order | **O(n²)** |
| **Space** | In-place sorting | **O(1)** |

### Best Case is O(n)!
When the array is already sorted, the `while` condition `arr[j] > key` is always False immediately. So the inner while loop never runs, and we only do the outer loop once per element → **O(n)**.

---

## 🚨 10. Common Beginner Mistakes

| Mistake | Problem | Fix |
|---------|---------|-----|
| Starting outer loop from `i=0` | `j = i-1 = -1` — wrong! | Start from `i = 1` |
| `arr[j] >= key` in while | Swaps equal elements unnecessarily; breaks stability for some cases | Use `arr[j] > key` for ascending sort |
| `j = i` instead of `j = i-1` | Compares key with itself first | Set `j = i - 1` |
| Not placing `arr[j+1] = key` OUTSIDE the while loop | Key is never inserted | Always insert AFTER the while loop |

---

## 🎤 12. Interview Questions

| Question | Answer |
|----------|--------|
| Is Insertion Sort stable? | **Yes** — equal elements maintain relative order |
| Best use case? | Small arrays or nearly sorted data |
| Used in practice? | Yes! Python's Timsort uses Insertion Sort for small subarrays |
| Compare with Bubble Sort? | Both O(n²) but Insertion Sort makes fewer comparisons on average |

# 🔀 Algorithm 6: Merge Sort

---

## 1️⃣ Introduction

### What is Merge Sort?
Merge Sort uses the powerful **"Divide and Conquer"** strategy:
1. **Divide** the array into two halves
2. **Recursively sort** each half
3. **Merge** the two sorted halves back together

### Real-Life Analogy 📚
Imagine you have two stacks of cards, both already sorted (face up, smallest on top). To merge them into one sorted stack:
- Compare the top cards of both stacks.
- Take the smaller card and place it on the result stack.
- Repeat until one stack is empty, then place all remaining cards from the other stack.

Merge Sort first **breaks everything down** into individual cards (which are trivially sorted), then **merges them back up**.

### Where Is It Used?
- **External sorting** (sorting data too big for RAM, stored on disk)
- Sorting **linked lists** (where random access is slow)
- Merge Sort is the basis of **Python's Timsort**

---

## 2️⃣ Problem Understanding

```
Input:  [38, 27, 43, 3, 9, 82, 10]
Output: [3, 9, 10, 27, 38, 43, 82]
```

---

## 3️⃣ Core Intuition

> **"A list of 1 element is already sorted. Merge two sorted lists into one sorted list."**

Merge Sort trusts recursion to sort each half. Our job is just to **merge** correctly.

---

## 4️⃣ Dry Run Example — Recursion Tree

**Array = [38, 27, 43, 3]**

```
                    [38, 27, 43, 3]
                   DIVIDE at mid=2
                 /                 \
          [38, 27]               [43, 3]
         DIVIDE at 1           DIVIDE at 1
         /       \             /         \
       [38]      [27]        [43]         [3]
       (sorted)  (sorted)   (sorted)    (sorted)
         \       /              \        /
          MERGE                  MERGE
         [27, 38]               [3, 43]
               \                /
                    MERGE
               [3, 27, 38, 43]  ✅
```

### How Does the MERGE Work?

Merging [27, 38] and [3, 43]:
```
Left  = [27, 38]   i=0
Right = [3,  43]   j=0
Result = []

Step 1: left[0]=27 vs right[0]=3  → 3 < 27 → take 3  → result=[3],   j=1
Step 2: left[0]=27 vs right[1]=43 → 27 < 43 → take 27 → result=[3,27], i=1
Step 3: left[1]=38 vs right[1]=43 → 38 < 43 → take 38 → result=[3,27,38], i=2
Step 4: left exhausted → append rest of right: [43]
Final: [3, 27, 38, 43] ✅
```

---

## 5️⃣ Step-by-Step Algorithm

```
MERGE SORT FUNCTION:
Step 1: If array has 0 or 1 elements → return it (already sorted)
Step 2: Find mid = len(array) // 2
Step 3: left  = merge_sort(array[:mid])   → sort left half recursively
Step 4: right = merge_sort(array[mid:])   → sort right half recursively
Step 5: return merge(left, right)          → merge the two sorted halves

MERGE FUNCTION:
Step 1: Create an empty result list
Step 2: Use two pointers i (for left) and j (for right), both start at 0
Step 3: While both pointers are within bounds:
          If left[i] <= right[j]: add left[i] to result, i++
          Else:                   add right[j] to result, j++
Step 4: Add remaining elements from left (if any)
Step 5: Add remaining elements from right (if any)
Step 6: Return result
```


````python
# ============================================================
# MERGE SORT — Full Python Code (From Scratch)
# ============================================================

def merge(left, right):
    """
    Merges two sorted lists (left and right) into one sorted list.
    This is the CORE of Merge Sort.
    """
    result = []   # The merged result will be built here
    i = 0         # Pointer for left list
    j = 0         # Pointer for right list
    
    # Compare elements from both lists and take the smaller one
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:       # Left element is smaller or equal
            result.append(left[i])
            i += 1
        else:                          # Right element is smaller
            result.append(right[j])
            j += 1
    
    # If any elements remain in left, add them all
    result.extend(left[i:])
    
    # If any elements remain in right, add them all
    result.extend(right[j:])
    
    return result


def merge_sort(arr):
    """
    Recursively splits array in half, sorts each half,
    then merges the sorted halves.
    """
    
    # BASE CASE: A list with 0 or 1 elements is already sorted
    if len(arr) <= 1:
        return arr
    
    # DIVIDE: Find the midpoint
    mid = len(arr) // 2
    
    # CONQUER: Recursively sort each half
    left  = merge_sort(arr[:mid])    # Sort the left half
    right = merge_sort(arr[mid:])    # Sort the right half
    
    # COMBINE: Merge the two sorted halves
    return merge(left, right)


# ─── Verbose version showing recursion depth ──────────────

depth = 0

def merge_sort_verbose(arr, depth=0):
    indent = "  " * depth
    print(f"{indent}merge_sort({arr})")
    
    if len(arr) <= 1:
        print(f"{indent}→ Base case, return {arr}")
        return arr
    
    mid = len(arr) // 2
    left  = merge_sort_verbose(arr[:mid], depth + 1)
    right = merge_sort_verbose(arr[mid:], depth + 1)
    
    result = merge(left, right)
    print(f"{indent}merge({left}, {right}) → {result}")
    return result


# ─── Test ─────────────────────────────────────────────────

print("=" * 55)
print("MERGE SORT — Recursion Tree Visualization")
print("=" * 55)
result = merge_sort_verbose([38, 27, 43, 3])
print(f"\nFinal sorted: {result}")

print("\n" + "=" * 55)

# Larger test
arr = [38, 27, 43, 3, 9, 82, 10]
print(f"\nLarger test:")
print(f"Input:  {arr}")
print(f"Output: {merge_sort(arr)}")
````

## 7️⃣ Line-by-Line Code Explanation

### `merge_sort` Function:
| Line | Explanation |
|------|-------------|
| `if len(arr) <= 1: return arr` | Base case: 0 or 1 element is already sorted |
| `mid = len(arr) // 2` | Find the midpoint for splitting |
| `left = merge_sort(arr[:mid])` | Recursively sort the LEFT half |
| `right = merge_sort(arr[mid:])` | Recursively sort the RIGHT half |
| `return merge(left, right)` | Merge sorted halves and return |

### `merge` Function:
| Line | Explanation |
|------|-------------|
| `result = []` | Empty list to build the merged result |
| `while i < len(left) and j < len(right)` | Compare while both have elements |
| `if left[i] <= right[j]` | Take from left if smaller or equal |
| `result.extend(left[i:])` | Dump remaining left elements |
| `result.extend(right[j:])` | Dump remaining right elements |

---

```

                SPLIT at mid=3
              /                             \
    [38, 27, 43]                         [3, 9, 82, 10]

 [38]        [27, 43]              [3, 9]           [82, 10]

```

## 9️⃣ Complexity Analysis

| Case | Time Complexity | Space Complexity |
|------|-----------------|------------------|
| **Best Case** | O(n log n) | O(n) |
| **Average Case** | O(n log n) | O(n) |
| **Worst Case** | O(n log n) | O(n) |

### Why O(n log n)?
- The array is split log₂(n) times (depth of recursion tree)
- At each level, we do O(n) work to merge
- Total = **n × log n**

### Why O(n) space?
Merge creates **new lists** at each level. The total extra memory used is proportional to the size of the original array.

## 🚨 10. Common Beginner Mistakes

| Mistake | Effect | Fix |
|---------|--------|-----|
| `arr[mid:]` vs `arr[mid+1:]` | Missing one element OR wrong split | Always use `arr[:mid]` and `arr[mid:]` |
| Forgetting base case | **Infinite recursion** / Stack overflow | `if len(arr) <= 1: return arr` |
| Merging in-place (modifying original) | Incorrect results | Always build a NEW result list in merge |
| `left[i] < right[j]` (strict less) | Works but breaks stability | Use `<=` to preserve relative order (stability) |

| Is it in-place? | **No** — requires O(n) extra space |
| What is the time complexity? | Always **O(n log n)** — consistent! |
| Advantage over Quick Sort? | Guaranteed O(n log n); great for linked lists |
| Disadvantage? | Requires O(n) extra space |

# ⚡ Algorithm 7: Quick Sort

---

## 1️⃣ Introduction

### What is Quick Sort?
Quick Sort is another **Divide and Conquer** algorithm, and is one of the **fastest sorting algorithms in practice**. It works by selecting a **pivot** element and **partitioning** the array around it.

### Real-Life Analogy 🗂️
Imagine sorting files by year:
1. Pick any file as the "pivot" (say, year 2020).
2. Put all files from **before 2020** on the LEFT.
3. Put all files from **after 2020** on the RIGHT.
4. Now sort the left and right piles the same way (recursively).

The pivot file is already in its **correct final position**!

### Where Is It Used?
- C's `qsort()` library function
- Java's `Arrays.sort()` for primitive types
- Used in most systems that need fast, in-place sorting

---

## 2️⃣ Core Intuition

> **"Place one element (the pivot) in its EXACT correct position. All elements to its left are smaller, all to its right are larger. Then recursively sort left and right."**

---

## 3️⃣ Dry Run — Partition Step (Most Important!)

**Array = [3, 6, 8, 10, 1, 2, 1], pivot = last element = 1**

```
Array = [3, 6, 8, 10, 1, 2, 1]
pivot = arr[high] = 1
i = -1   (boundary between "≤ pivot" and "> pivot")

j=0: arr[0]=3  > pivot(1) → no action
j=1: arr[1]=6  > pivot(1) → no action
j=2: arr[2]=8  > pivot(1) → no action
j=3: arr[3]=10 > pivot(1) → no action
j=4: arr[4]=1  ≤ pivot(1) → i++ → i=0 → swap arr[0] and arr[4]
     Array = [1, 6, 8, 10, 3, 2, 1]
j=5: arr[5]=2  > pivot(1) → no action

End of loop: swap arr[i+1]=arr[1] with arr[high]=arr[6] (the pivot)
     Array = [1, 1, 8, 10, 3, 2, 6]
pivot_index = i+1 = 1

arr[1] = 1 is now in its CORRECT position!
Left part  = [1]         → all ≤ 1 ✅
Right part = [8,10,3,2,6] → all > 1 ✅
```

**Simpler example — Array = [4, 2, 7, 1, 5], pivot = 5 (last)**

```
i = -1, pivot = 5
j=0: arr[0]=4 ≤ 5 → i=0 → swap arr[0]↔arr[0] → [4, 2, 7, 1, 5]
j=1: arr[1]=2 ≤ 5 → i=1 → swap arr[1]↔arr[1] → [4, 2, 7, 1, 5]
j=2: arr[2]=7 > 5 → skip
j=3: arr[3]=1 ≤ 5 → i=2 → swap arr[2]↔arr[3] → [4, 2, 1, 7, 5]
End: swap arr[i+1]=arr[3] with arr[4] (pivot)
     [4, 2, 1, 5, 7]
pivot is now at index 3. ✅ Elements < 5 on left, > 5 on right.
```

---

## 5️⃣ Step-by-Step Algorithm

```
QUICK SORT FUNCTION:
Step 1: If low >= high → return (base case: 0 or 1 element)
Step 2: Call partition(arr, low, high) → returns pivot_index
Step 3: Recursively sort left part:  quick_sort(arr, low, pivot_index - 1)
Step 4: Recursively sort right part: quick_sort(arr, pivot_index + 1, high)

PARTITION FUNCTION (using last element as pivot):
Step 1: pivot = arr[high]   (last element)
Step 2: i = low - 1         (boundary: everything ≤ i is ≤ pivot)
Step 3: For j from low to high-1:
Step 4:   If arr[j] <= pivot:
Step 5:     i += 1
Step 6:     Swap arr[i] with arr[j]
Step 7: Swap arr[i+1] with arr[high]  (place pivot in correct position)
Step 8: Return i + 1                  (the pivot's final index)
```


````python
# ============================================================
# QUICK SORT — Full Python Code (From Scratch)
# ============================================================

def partition(arr, low, high):
    """
    Partitions arr[low..high] around pivot = arr[high].
    After partition:
      - Elements at indices [low .. pivot_idx-1] are <= pivot
      - Element at pivot_idx IS the pivot (in its correct position)
      - Elements at indices [pivot_idx+1 .. high] are > pivot
    Returns: the final index of the pivot element.
    """
    
    pivot = arr[high]    # Choose the LAST element as pivot
    i = low - 1          # i is the boundary: arr[low..i] contains elements <= pivot
    
    for j in range(low, high):   # j scans from low to high-1
        
        if arr[j] <= pivot:       # Current element belongs in the "left" (≤ pivot) section
            i += 1                # Expand the left section boundary
            arr[i], arr[j] = arr[j], arr[i]   # Swap to move element to left section
    
    # Place pivot in its correct position (just after the left section)
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    
    return i + 1    # Return the pivot's final index


def quick_sort(arr, low, high):
    """
    Recursively sorts arr[low..high] using Quick Sort.
    Call as: quick_sort(arr, 0, len(arr) - 1)
    """
    
    if low < high:    # BASE CASE: if low >= high, nothing to sort
        
        # Get pivot's correct position (partition the array)
        pivot_idx = partition(arr, low, high)
        
        # Recursively sort LEFT side of pivot
        quick_sort(arr, low, pivot_idx - 1)
        
        # Recursively sort RIGHT side of pivot
        quick_sort(arr, pivot_idx + 1, high)


# ─── Verbose partition ────────────────────────────────────

def partition_verbose(arr, low, high):
    pivot = arr[high]
    i = low - 1
    
    print(f"  Pivot = {pivot}")
    print(f"  Array before partition: {arr[low:high+1]}")
    
    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
            print(f"  arr[{j}]={arr[j]} ≤ {pivot} → swap → {arr[low:high+1]}")
        else:
            print(f"  arr[{j}]={arr[j]} > {pivot} → no swap")
    
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    print(f"  Place pivot at index {i+1} → {arr[low:high+1]}")
    return i + 1


# ─── Test ─────────────────────────────────────────────────

# Simple clean test
arr1 = [4, 2, 7, 1, 5, 3]
print(f"Input:  {arr1}")
quick_sort(arr1, 0, len(arr1) - 1)
print(f"Output: {arr1}")

print()

# Verbose partition demo
arr2 = [4, 2, 7, 1, 5]
print("Partition demo on:", arr2)
pivot_pos = partition_verbose(arr2, 0, len(arr2) - 1)
print(f"Pivot index: {pivot_pos}, Pivot value: {arr2[pivot_pos]}")
print(f"Left of pivot:  {arr2[:pivot_pos]}")
print(f"Right of pivot: {arr2[pivot_pos+1:]}")

print()

# Edge cases
single = [1]
quick_sort(single, 0, len(single) - 1)
print(f"Single element: {single}")

already = [1, 2, 3, 4, 5]
quick_sort(already, 0, len(already) - 1)
print(f"Already sorted: {already}")

reverse = [5, 4, 3, 2, 1]
quick_sort(reverse, 0, len(reverse) - 1)
print(f"Reverse sorted: {reverse}")
````

## 9️⃣ Complexity Analysis

| Case | Situation | Time Complexity | Space |
|------|-----------|-----------------|-------|
| **Best Case** | Pivot always at the middle | O(n log n) | O(log n) |
| **Average Case** | Random array | O(n log n) | O(log n) |
| **Worst Case** | Sorted/Reverse sorted + last pivot | **O(n²)** | O(n) |
| **In-place?** | Yes — no extra array | — | O(log n) stack |

### Worst Case Explained:
If the array is `[1, 2, 3, 4, 5]` and we always pick the last element as pivot:
- Round 1: pivot=5, left=[1,2,3,4], right=[]  → completely unbalanced!
- Round 2: pivot=4, left=[1,2,3], right=[] → still unbalanced!
- This leads to **n** recursive calls each doing **n** work → **O(n²)**

**Fix:** Use **Random Pivot** or **Median-of-Three** pivot selection.

## 🚨 10. Common Beginner Mistakes

| Mistake | Effect | Fix |
|---------|--------|-----|
| `for j in range(low, high+1)` | Compares pivot with itself | Use `range(low, high)` — stop BEFORE high |
| Wrong base case: `if low == high` | Misses the case `low > high` | Use `if low < high` |
| `return i` instead of `return i+1` | Puts pivot at wrong index | Always return `i + 1` |
| Forgetting to place pivot at end | Pivot is in middle of partition, not at boundary | Always do the final swap after the loop |

## 🤔 When To Use Which Algorithm?

| Situation | Best Choice | Why |
|-----------|-------------|-----|
| Unsorted list, find element | **Linear Search** | No sorting needed |
| Sorted list, find element | **Binary Search** | O(log n) vs O(n) |
| Small array, simple sort | **Insertion Sort** | Low overhead, stable |
| Nearly sorted array | **Insertion Sort** | O(n) best case |
| Need guaranteed O(n log n) | **Merge Sort** | Always consistent |
| General purpose, fast | **Quick Sort** | Best average cache performance |
| Memory is limited | **Quick Sort** or **Heap Sort** | In-place sorting |
| Stability required | **Merge Sort** or **Insertion Sort** | Maintains relative order |
| External sorting (disk) | **Merge Sort** | Sequential access pattern |

## 🎤 Top Interview Tips

1. **Always clarify** if the array is sorted before choosing a search algorithm.
2. **Binary Search** can be applied to problems other than arrays — search on **answer space**!
3. **Stability** means equal elements keep their original relative order. Important for multi-key sorting.
4. **In-place** means O(1) extra space. Quick Sort is in-place; Merge Sort is not.
5. **Quick Sort worst case** (O(n²)) is avoided in practice by using random pivot.
6. **Timsort** (Python's built-in sort) = hybrid of Merge Sort + Insertion Sort → O(n log n) guaranteed.
7. For interviews: **know Merge Sort and Quick Sort deeply** — they appear in many problems.

````python
# ============================================================
# FINAL VERIFICATION — All Algorithms Working Correctly
# ============================================================

print("=" * 60)
print("FINAL VERIFICATION OF ALL ALGORITHMS")
print("=" * 60)

test_array_unsorted = [64, 34, 25, 12, 22, 11, 90]
test_array_sorted   = [11, 12, 22, 25, 34, 64, 90]


# ── Linear Search ─────────────────────────────────────────
def linear_search(arr, target):
    for i in range(len(arr)):
        if arr[i] == target:
            return i
    return -1

idx = linear_search(test_array_unsorted, 22)
print(f"\n1. Linear Search  → Found 22 at index: {idx}")  # Expected: 4


# ── Binary Search ─────────────────────────────────────────
def binary_search_iterative(arr, target):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = left + (right - left) // 2
        if arr[mid] == target:   return mid
        elif arr[mid] < target:  left = mid + 1
        else:                    right = mid - 1
    return -1

idx = binary_search_iterative(test_array_sorted, 22)
print(f"2. Binary Search   → Found 22 at index: {idx}")  # Expected: 2


# ── Bubble Sort ───────────────────────────────────────────
def bubble_sort(arr):
    arr = arr[:]
    n = len(arr)
    for i in range(n - 1):
        swapped = False
        for j in range(n - 1 - i):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr

print(f"3. Bubble Sort     → {bubble_sort(test_array_unsorted)}")


# ── Selection Sort ────────────────────────────────────────
def selection_sort(arr):
    arr = arr[:]
    n = len(arr)
    for i in range(n - 1):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

print(f"4. Selection Sort  → {selection_sort(test_array_unsorted)}")


# ── Insertion Sort ────────────────────────────────────────
def insertion_sort(arr):
    arr = arr[:]
    n = len(arr)
    for i in range(1, n):
        key = arr[i]
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1
        arr[j + 1] = key
    return arr

print(f"5. Insertion Sort  → {insertion_sort(test_array_unsorted)}")


# ── Merge Sort ────────────────────────────────────────────
def merge(left, right):
    result, i, j = [], 0, 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i]); i += 1
        else:
            result.append(right[j]); j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    return merge(merge_sort(arr[:mid]), merge_sort(arr[mid:]))

print(f"6. Merge Sort      → {merge_sort(test_array_unsorted)}")


# ── Quick Sort ────────────────────────────────────────────
def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1
    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]
    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1

def quick_sort(arr, low, high):
    if low < high:
        pivot_idx = partition(arr, low, high)
        quick_sort(arr, low, pivot_idx - 1)
        quick_sort(arr, pivot_idx + 1, high)

qs_arr = test_array_unsorted[:]
quick_sort(qs_arr, 0, len(qs_arr) - 1)
print(f"7. Quick Sort      → {qs_arr}")

print()
print("=" * 60)
print("✅ All algorithms verified error-free!")
print("=" * 60)
print()
print("Expected for all sort outputs: [11, 12, 22, 25, 34, 64, 90]")
````

## 📌 Quick Revision Cheatsheet

```
SEARCHING:
  Linear Search  → O(n)     → works on unsorted arrays
  Binary Search  → O(log n) → requires sorted array, uses left/mid/right pointers

SORTING (Simple — O(n²)):
  Bubble Sort    → Compare adjacent, swap if wrong order. Optimized with 'swapped' flag.
  Selection Sort → Find minimum in unsorted part, place at correct position.
  Insertion Sort → Take each element, insert into correct position in sorted left portion.

SORTING (Efficient — O(n log n)):
  Merge Sort  → Divide in half → sort each half → MERGE. Stable, O(n) space.
  Quick Sort  → Pick pivot → PARTITION (left ≤ pivot, right > pivot) → sort each side.
                In-place, unstable, O(n²) worst case.
```
