# **Dijkstra's Algorithm**
---

1. Start at source node → distance = 0
2. All other nodes → distance = INFINITY (unknown)
3. Always visit the UNVISITED node with SMALLEST distance
4. Update neighbors if we found a SHORTER path
5. Repeat until all nodes visited

```python
def dijkstra(graph, src):
    """
    graph : adjacency matrix → graph[i][j] = weight (0 = no edge)
    src   : starting node
    """
    n = len(graph)

    # Step 1: Initialize distances and visited
    dist    = []
    visited = []
    parent  = []

    for i in range(n):
        dist.append(99999)    # infinity = unknown distance
        visited.append(False) # no node visited yet
        parent.append(-1)     # no parent yet

    dist[src] = 0             # distance to source = 0

    print(f"\nStarting from Node {src}")
    print(f"\n{'Step':<6} {'Visited':<8} {'Distances'}")
    print("-" * 40)


    # Step 2: Process all nodes one by one
    for step in range(n):

        # Step 2a: Find unvisited node with minimum distance (manual search)
        u = -1
        for v in range(n):
            if not visited[v]:
                if u == -1 or dist[v] < dist[u]:
                    u = v

        if dist[u] == 99999:  # remaining nodes are unreachable
            break

        visited[u] = True     # mark as visited

        # Print current state
        dist_display = ""
        for d in dist:
            if d == 99999:
                dist_display += "∞  "
            else:
                dist_display += str(d) + "  "
        print(f"  {step+1:<5} Node {u:<3}  [{dist_display}]")

        # Step 2b: Update distances of neighbors
        for v in range(n):
            if graph[u][v] != 0 and not visited[v]: # there is an edge
                new_dist = dist[u] + graph[u][v] # distance through u

                if new_dist < dist[v]:  # found shorter path!
                    dist[v]   = new_dist
                    parent[v] = u

    return dist, parent


# Traceback: Print actual path from src to dest
def print_path(parent, src, dest):
    path = []
    current = dest

    while current != -1:  # go backwards from dest to src
        path.append(current)
        current = parent[current]

    # Reverse the path (it's backwards right now)
    left  = 0
    right = len(path) - 1
    while left < right:
        path[left], path[right] = path[right], path[left]
        left  += 1
        right -= 1

    print(" → ".join(str(node) for node in path))


# Print final results
def print_results(dist, parent, src):
    n = len(dist)
    print(f"\n--- Shortest Paths from Node {src} ---")
    print(f"{'Dest':<8} {'Distance':<12} {'Path'}")
    print("-" * 35)

    for i in range(n):
        if i == src:
            continue
        print(f"  {src} → {i}   ", end="")
        if dist[i] == 99999:
            print("Unreachable")
        else:
            print(f"{dist[i]:<10}  ", end="")
            print_path(parent, src, i)


# Test
# Adjacency Matrix (0 = no edge)
# Nodes: A=0, B=1, C=2, D=3

graph = [
#    A  B  C  D
    [0, 4, 2, 0],   # A connects to B(4), C(2)
    [4, 0, 0, 3],   # B connects to A(4), D(3)
    [2, 0, 0, 1],   # C connects to A(2), D(1)
    [0, 3, 1, 0],   # D connects to B(3), C(1)
]

src = 0 # Start from node A

dist, parent = dijkstra(graph, src)
print_results(dist, parent, src)
```
