# **📦 Basic Class The foundation — defining a class, constructor, attributes, and methods.**

```python
class Dog:
    # Class Variable (shared by all instances)
    species = "Canis Familiaris"
# Constructor (__init__) — runs when object is created
    def __init__(self, name, age, breed):
        # Instance Variables (unique to each object)
        self.name  = name
        self.age   = age
        self.breed = breed
# Instance Method
    def bark(self):
        return f"{self.name} says: Woof!"
# Method using instance variables
    def describe(self):
        return f"{self.name} is a {self.age} year old {self.breed}"
# __str__ — controls print() output
    def __str__(self):
        return f"Dog({self.name}, {self.breed})"
```

```python
# ✅ Creating Objects (Instances)
dog1 = Dog("Buddy", 3, "Golden Retriever")
dog2 = Dog("Max",   5, "German Shepherd")
# ✅ Accessing Attributes
print(dog1.name)       # Buddy
print(dog1.species)    # Canis Familiaris  (class variable)
# ✅ Calling Methods
print(dog1.bark())     # Buddy says: Woof!
print(dog2.describe()) # Max is a 5 year old German Shepherd
print(dog1)            # Dog(Buddy, Golden Retriever)
```

# **💡 Key Points**
- ✓
self refers to the current instance — always the 1st parameter
- ✓
Class variables are shared across all instances
- ✓
Instance variables are unique to each object
- ✓
__init__ is the constructor — called automatically on creation
- ✓
__str__ controls what print() displays

# **2. 🧬 Inheritance Child classes inherit attributes & methods from parent. Reuse + extend.**

```python
# ========== PARENT CLASS ==========
class Animal:
    def __init__(self, name, sound):
        self.name  = name
        self.sound = sound
        self.is_alive = True
    def speak(self):
        return f"{self.name} says {self.sound}!"
    def eat(self, food):
        return f"{self.name} is eating {food}"
# ========== CHILD CLASS ==========
class Cat(Animal):
    def __init__(self, name, breed, indoor=True):
        super().__init__(name, sound="Meow")   # call parent
        self.breed  = breed
        self.indoor = indoor
    def purr(self):                            # new method
        return f"{self.name} is purring…"
    def speak(self):                           # override parent
        return f"{self.name} ({self.breed}) says Meow softly~"
class Parrot(Animal):
    def __init__(self, name, vocabulary=None):
        super().__init__(name, sound="Squawk")
        self.vocabulary = vocabulary or []
    def learn_word(self, word):
        self.vocabulary.append(word)
        return f"{self.name} learned: '{word}'"
    def speak(self):
        if self.vocabulary:
            return f"{self.name} says: '{self.vocabulary[-1]}'"
        return super().speak()                 # fall back to parent
```

```python
# ✅ Usage
cat = Cat("Whiskers", "Persian")
print(cat.speak())          # Whiskers (Persian) says Meow softly~
print(cat.eat("fish"))      # Whiskers is eating fish  ← inherited!
print(cat.purr())           # Whiskers is purring…
parrot = Parrot("Polly")
print(parrot.speak())       # Polly says Squawk!
parrot.learn_word("Hello")
print(parrot.speak())       # Polly says: 'Hello'
# ✅ isinstance checks
print(isinstance(cat, Cat))     # True
print(isinstance(cat, Animal))  # True  ← Cat IS an Animal
```

# **💡 Key Points**
- ✓
Child class syntax: class Child(Parent)
- ✓
super().__init__() calls the parent constructor
- ✓
Child classes inherit ALL parent methods automatically
- ✓
Methods can be overridden in child classes
- ✓
isinstance() checks if an object belongs to a class or its parent
