# **Fraction Knapsack**
---

```python
def fra_knap(capacity, items):
    n = len(items)
    
    # calculate ratio
    ratio_item = []
    for i in range(n):
        value = items[i][0]
        weight = items[i][1]
        ratio = value / weight
        ratio_item.append([ratio, value, weight])

    # bubble sort
    for i in range(n):
        for j in range(0, n-i-1):
            if ratio_item[j][0] < ratio_item[j+1][0]:
                ratio_item[j], ratio_item[j+1] = ratio_item[j+1], ratio_item[j]

    # fill knapsack
    total_value = 0.0
    remining_capacity = capacity

    print("Item   Ratio    Taken Weight    Value Gained") 
    for i in range(n):
        ratio = ratio_item[i][0]
        value = ratio_item[i][1]
        weight = ratio_item[i][2]

        if remining_capacity == 0:
            break

        if weight <= remining_capacity:
            total_value += value
            remining_capacity -= weight
            print(f"  {i+1}      {ratio:.2f}       {weight}               {value:.2f}")
        else:
            fraction = remining_capacity / weight
            total_value += value * fraction
            print(f"  {i+1}      {ratio:.2f}       {remining_capacity}               {total_value:.2f}    fraction: {fraction:.2f}")
            remining_capacity = 0

    return total_value 


items = [[60, 10], [100, 20], [120, 30]]
capacity = 50
print(f"Knapsack Capacity : {capacity}")
print(f"Number of Items   : {len(items)}")

max_value = fra_knap(capacity, items)
print(f"\nMaximum Value in Knapsack = {max_value:.2f}")
```
