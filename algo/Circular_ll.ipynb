# **Create Circular linked list**

- Traversal
- Searching
- Length
- Insertion:
- Insert at the beginning
- Insert at the end
- Insert at a specific position
- **Deletion**:
1. - Delete from the beginning
2. - Delete from the end
3. - Delete a specific node

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None

class CircularLL:
    def __init__(self):
        self.tail = None
    

    def Traversal(self):
        if self.tail is None:          
            print("List is empty")
            return
        current = self.tail           
        while True:
            current = current.next
            print(current.data, end=" --> ")
            if current == self.tail:
                break
        print(current.next.data)       # prints head (tail.next) to show circularity


    def searchcll(self, key):
        if self.tail is None:
            print("List is empty")
            return
        current = self.tail
        while True:
            current = current.next
            if current.data == key:    #  was comparing search==key (always True!)
                                       #  must check current.data == key
                print(f"Found {key} at node {current}")
                return
            if current == self.tail:   #  was infinite loop, no exit condition
                break
        print("Not found")

    
    def insert_cll_beginning(self, value):
        new_node = Node(value)
        if self.tail is None:          # handle empty list
            self.tail = new_node
            new_node.next = new_node
            return
        new_node.next = self.tail.next #  point new node to current head
        self.tail.next = new_node      #           point tail to new node (new head)
        # tail stays the same, new_node is the new head
        self.Traversal()

    def insert_cll_end(self,value):
        new_node = Node(value)
        current = self.tail
        current.next = new_node
        current = new_node
        new_node.next = current

    def insert_cll_end(self, value):
        new_node = Node(value)
        if self.tail is None:          # handle empty list
            self.tail = new_node
            new_node.next = new_node
            return
        new_node.next = self.tail.next # new node points to head
        self.tail.next = new_node      #           old tail points to new node
        self.tail = new_node           #           update tail to new node
        self.Traversal()

    def insert_at_position(self, data, position):
        if self.tail is None:          # use self.tail not bare `tail`
            print("CLL is empty")
            return
        current = self.tail
        count = 1
        while True:
            current = current.next
            if count == position - 1: # stop at node BEFORE target position
                new_node = Node(data)
                new_node.next = current.next  # was setting new_node = current.next
                current.next = new_node       # (overwrote new_node instead of linking)
                return
            if current == self.tail:
                print("Position out of range")
                return
            count += 1


    def deletion_cll_beginning(self):
     if self.tail is None:                    # Empty list check
        print("List is empty")
        return
     if self.tail.next == self.tail:          # Only one node
        self.tail = None
        return
     self.tail.next = self.tail.next.next  
                                            #  skip head, tail now points to new head
     self.Traversal()

    def deletion_cll_end(self):
     if self.tail is None:                    # Empty list check
        print("List is empty")
        return
     if self.tail.next == self.tail:          # Only one node
        self.tail = None
        return
     current = self.tail
     while current.next != self.tail:         
        current = current.next               #traverse until node BEFORE tail
     current.next = self.tail.next    
     self.tail = current                      #point second-to-last to head, update tail
     self.Traversal()

    def deletion_at_position(self, position):
     if self.tail is None:
        print("List is empty")
        return

    # Case 1: Delete at Position 1 (Head)
     if position == 1:
        # If there's only one node
        if self.tail.next == self.tail:
            self.tail = None
        else:
            # Point tail to the node after the current head
            self.tail.next = self.tail.next.next
        self.Traversal()
        return

    # Delete at Position k
    # We start 'prev' at head and 'current' at head.next
     prev = self.tail.next
     current = prev.next
     count = 2 # Since we are already looking at the second node
    
    # We loop until we complete a circle back to the head
     while prev.next != self.tail.next:
        if count == position:
            # Check if we are deleting the tail node
            if current == self.tail:
                prev.next = current.next
                self.tail = prev
            else:
                prev.next = current.next
            
            self.Traversal()
            return
        
        prev = current
        current = current.next
        count += 1

     print("Position out of range")

    def length(self):
        if self.tail is None:
            print("List is empty")
            return 0
        current = self.tail
        count = 1                                # Start at 1 to count tail itself
        while current.next != self.tail:
            current = current.next
            count += 1
        print(f"Length of CLL: {count}")
        return count
    
            
    
    
```

```python
cll = CircularLL()
cll.tail = Node(5)
node2 = Node(9)
node3 = Node(6)

cll.tail.next = node2
node2.next = node3
node3.next = cll.tail
cll.tail = node3  # tail  last node

cll.Traversal()

#insert at begning
print("#-------------------------------------------#")
print("Insert value at begning :")
cll.insert_cll_beginning(7)

#insert at end
print("#-------------------------------------------#")
print("Insert value at end :")
cll.insert_cll_end(8)

#insert at end
print("#-------------------------------------------#")
print(f"Insert 3 at 3 :")
cll.insert_at_position(3,3)
cll.Traversal()

#searching
print("#-------------------------------------------#")
print("Searching value:")
cll.searchcll(8)

#length
print("#-------------------------------------------#")
print("length of linked list :")
cll.length()


#delation at begning 
print("#-------------------------------------------#")
print("delete circular linked list begning node :")
cll.deletion_cll_beginning()

#delation at begning 
print("#-------------------------------------------#")
print("delete circular linked list end node :")
cll.deletion_cll_end()

#delation at position
print("#-------------------------------------------#")
print("delete circular linked list at given positon :")
cll.deletion_at_position(2)
```
